## 학습 개요

1. **학습 주제**
  - **토큰화(Tokenization)**: 텍스트를 토큰 단위로 분할하고 정수 ID로 변환하는 과정
  - **임베딩(Embedding)**: 토큰을 고정 차원의 연속적인 벡터 공간으로 매핑하는 기법
  - **시퀀스 모델 발전사**: RNN부터 Transformer까지 자연어 처리 모델의 진화 과정 이해

2. **학습 목표**
  - 토크나이저가 무엇이고 토큰화가 무엇인지에 대해서 **설명**할 수 있다.
  - 토큰화를 왜 하는지에 대해서 **설명**할 수 있다.
  - 토큰화된 토큰들을 임베딩 벡터로 **변환**할 수 있다.
  - 임베딩 벡터를 이용하여 어떤 식으로 활용할 수 있는지 **설명**할 수 있다.
  - RNN/LSTM의 순차 처리 방식과 한계를 **이해**할 수 있다.
  - Attention 메커니즘의 작동 원리와 효과를 **설명**할 수 있다.
  - Encoder-Decoder 구조를 활용한 번역 태스크를 **구현**할 수 있다.
  - Transformer 아키텍처의 Self-Attention 메커니즘을 **이해**할 수 있다.

3. **핵심 개념**
  - **토큰화(Tokenization)**: 자연어 텍스트를 모델이 처리할 수 있는 토큰 단위로 분할하고, 각 토큰을 고유한 정수 ID로 매핑하는 전처리 과정
  - **임베딩 벡터(Embedding Vector)**: 토큰 ID를 고정 차원의 실수 벡터로 변환하여 단어 간 의미적 유사도를 학습 가능하게 만드는 표현 방식
  - **RNN(Recurrent Neural Network)**: 이전 시점의 은닉 상태를 현재 시점 입력과 결합하여 순차 데이터를 처리하는 신경망 구조
  - **LSTM(Long Short-Term Memory)**: 장기 의존성 문제를 해결하기 위해 게이트 메커니즘을 도입한 RNN의 개선 버전
  - **Attention 메커니즘**: 입력 시퀀스의 각 위치에 대한 가중치를 동적으로 계산하여 중요한 정보에 집중하는 기법
  - **Encoder-Decoder**: 입력 시퀀스를 고정 길이 문맥 벡터로 인코딩한 후, 이를 디코딩하여 출력 시퀀스를 생성하는 구조
  - **Transformer**: Self-Attention을 핵심으로 하여 병렬 처리가 가능하고 장거리 의존성을 효과적으로 학습하는 최신 아키텍처

4. **선행 지식**
  - 파이썬 기초 문법(함수, 클래스, 타입 힌트)
  - PyTorch 기본 사용법(Tensor, nn.Module, forward)
  - 신경망 기초 개념(가중치, 활성화 함수, 순전파)
  - 자연어 처리 기초 개념(단어, 문장, 어휘 사전)

학습이 되어 있지 않은 빈 tokenizer를 생성합니다. 여기서는 BertWordPieceTokenizer를 불러옵니다.

**파라미터:**
- `strip_accents` : 입력 텍스트의 악센트(액센트)를 제거할지 여부를 결정하는 옵션입니다. 한국어는 라틴 문자처럼 café, résumé 와 같은 악센트 문자가 없기 때문에 `False`로 설정합니다.
- `lowercase` : 영어를 모두 소문자로 바꿉니다. `False`로 설정하면 영어를 대문자로 유지합니다.

````python
from tokenizers import BertWordPieceTokenizer

# 빈 tokenizer 생성 : vocabulary_size = 0 인 것을 확인하실 수 있습니다.
tokenizer = BertWordPieceTokenizer(
    lowercase=False,
    strip_accents=False,
)
tokenizer
````

아래 코드를 실행하여 토크나이저를 학습합니다.

**파라미터 설명:**
- `data_file` : 데이터 경로를 지정해줍니다. list 형태로 여러개의 파일을 지정해줄수도 있습니다.
- `vocab_size (default: 30000)` : 단어사전 크기를 지정할 수 있습니다. 어떠한 값이 가장 좋다는 것은 없지만, 값이 클수록 많은 단어의 의미를 담을 수 있습니다.
- `initial_alphabet` : 꼭 포함됐으면 하는 initial alphabet을 학습 전에 추가해줍니다.
    - initial은 학습하기 이전에 미리 단어를 vocab에 넣는 것을 의미합니다.
    - special token들도 initial에 vocab에 추가됩니다.
- `limit_alphabet (default: 1000)` : initial tokens의 갯수를 제한합니다.
- `min_frequency (default: 2)` : 최소 빈도수를 의미합니다. 만약 어떤 단어가 1번 나오면 vocab에 추가하지 않습니다.
- `special_tokens` : 특수 토큰을 넣을 수 있습니다.. BERT에는 다음과 같은 토큰이 들어가야 합니다.
    - `[PAD]` : 패딩을 위한 토큰
    - `[UNK]` : OOV(Out Of Vocabulary) 단어를 위한 토큰
    - `[CLS]` : 문장의 시작을 알리고 분류 문제에 사용되는 토큰
    - `[SEP]` : 문장 사이사이를 구별해주는 토큰
    - `[MASK]` : MLM(Masked Language Modeling) 태스크를 위한 마스크 토큰
- `wordpiece_prefix(default: '##')` : sub-word라는 것을 알려주는 표시입니다.
    - BERT는 기본적으로 '##'을 씁니다.
    - 예를 들어, `SS, ##AF, ##Y` 처럼 sub-word를 구분하기 위해 '##'을 사용합니다.
- `show_progress` : 학습 과정을 보여줍니다.

````python
data_file = 'naver_review.txt'
vocab_size = 30000        # 만들고 싶은 단어장 크기 (목표치이지, 반드시 정확히 이 수가 되는 건 아닙니다)
min_frequency = 2         # 2번 미만 등장한 후보는 단어장에 넣지 않음
initial_alphabet = []     # 반드시 포함할 글자를 미리 지정 (비워두면 데이터에서 자동 수집)
limit_alphabet = 6000     # 기본 글자(alphabet) 개수 상한.
                          # 한글은 완성형 글자가 1만 자가 넘어서 넉넉히 6000으로 잡았습니다.
                          # 기본값 1000으로 두면 한글 상당수가 [UNK]가 됩니다!
special_tokens = ["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"]
wordpieces_prefix = "##"  # 단어의 '중간 조각'임을 표시하는 접두사
show_progress=True

tokenizer.train(
    files=data_file,
    vocab_size=vocab_size,
    min_frequency=min_frequency,
    initial_alphabet=initial_alphabet,
    limit_alphabet=limit_alphabet,
    special_tokens=special_tokens,
    wordpieces_prefix=wordpieces_prefix,
    show_progress=True,
)

vocab = tokenizer.get_vocab()
print("vocab size : ", len(vocab))
print(sorted(vocab, key=lambda x: vocab[x])[:20])

# [확인 포인트] 출력된 앞 20개를 보세요.
#   - 0~4번은 우리가 지정한 special token들입니다. (순서도 지정한 그대로)
#   - 그 뒤로는 데이터에서 수집한 '한 글자' 토큰들이 나옵니다.
#   - 실제 len(vocab)이 30000보다 작을 수 있습니다. min_frequency 조건을 통과한
#     후보가 부족하면 목표치를 못 채우기 때문입니다. 이건 정상입니다.
````

````python
text = "I'm a student of SSAFY!"

encoded = tokenizer.encode(text)
print('🌱토큰화 결과:', encoded.tokens)
print('🌱정수 인코딩:', encoded.ids)
print('🌈디코딩 (정수로 인코딩된 토큰을 다시 문자열로 변환) :\n\t', tokenizer.decode(encoded.ids))

# [비교 실험] 같은 토크나이저에 한국어를 넣어보세요. 결과가 확연히 다릅니다.
print()
korean = "이 영화 정말 재미있었어요"
enc_ko = tokenizer.encode(korean)
print('🇰🇷 한국어 토큰화:', enc_ko.tokens)
print(f'   토큰 개수 - 영어 {len(encoded.tokens)}개 vs 한국어 {len(enc_ko.tokens)}개')
````

> 🔍 **출력에서 반드시 확인할 3가지**
>
> **① 앞뒤에 `[CLS]` 와 `[SEP]` 이 자동으로 붙었습니다**
> 우리가 넣지 않았는데 생겼죠. `BertWordPieceTokenizer` 가 BERT 규격에 맞춰 자동으로 붙여줍니다.
> 그래서 토큰 개수가 원문보다 2개 많습니다.
>
> **② 영어가 잘게 부서졌습니다**
> `SSAFY` 같은 단어는 `SS`, `##AF`, `##Y` 처럼 쪼개지거나, 심하면 글자 단위까지 내려갑니다.
> **`##` 은 "앞 토큰에 이어 붙는 조각"이라는 표시**입니다. 디코딩할 때 이 표시를 보고 다시 합칩니다.
>
> **③ 한국어는 훨씬 적은 토큰으로 표현됩니다**
> 한국어 영화 리뷰로 학습했으니 당연합니다. **토크나이저는 학습 데이터를 닮습니다.**
>
> 💡 **실무 인사이트 — 이게 왜 돈 문제인가**
> LLM API는 **토큰 단위로 과금**합니다. 같은 의미의 문장이라도 토크나이저에 따라 토큰 수가 2~3배 차이 납니다.
> 실제로 초기 GPT 모델들은 영어 위주로 학습된 토크나이저를 써서, **한국어가 영어보다 2~3배 비쌌습니다.**
> (한글 한 글자가 토큰 2~3개로 쪼개졌기 때문입니다)
> 최근 모델들은 다국어 토크나이저를 써서 격차가 많이 줄었습니다.


그렇다면 모델의 입력으로 넣기 위해서는 어떤 방식으로 토큰화해야 할까요? 위에 대한 답변은 앞으로 실습 코드를 진행하면서 나오기 때문에 이 점을 잊지 말고 계속 따라가시기 바랍니다.

### Concept Check
**토큰의 벡터화**: 각 토큰은 고유한 정수 ID로 표현되며, 이 값 자체에는 의미 정보가 없습니다. 신경망 모델이 의미를 학습할 수 있도록, 각 토큰을 N차원의 연속적인 임베딩 벡터로 변환합니다. 이를 수행할 수 있는 방법은 torch의 **`nn.Embedding`** 모듈을 사용하여 임베딩 벡터를 생성하는 것입니다. [API Reference](https://docs.pytorch.org/docs/stable/generated/torch.nn.Embedding.html)를 살펴보고 오세요.

1. `num_embeddings`: 임베딩 사전의 크기 (size of the dictionary of embeddings)
   - 여기서 `num_embeddings`는 고유한 토큰(단어, 문자 등)의 총 개수를 의미합니다. 즉, 어떤 `인덱스 → 벡터` 매핑 테이블을 만들 건데, 그 테이블에 몇 개의 항목이 들어가야 하는지를 정의하는 값입니다. tokenizer를 만들때 `vocab_size`와 동일한 값을 의미합니다.

2. `embedding_dim`: 각 임베딩 벡터의 차원 (the size of each embedding vector)
   - `embedding_dim`은 각 단어(또는 토큰)가 표현되는 벡터의 길이입니다. 즉, 하나의 단어를 어떤 숫자 벡터로 나타낼 때 그 벡터가 몇 차원인지 정하는 값입니다. 보통 `768`, `1024`, `2048` 같은 값을 사용합니다. ("어떤 값이 정답이다" 하는 값이 있는 건 아닙니다.)
   - 📌 자주 나오는 오해 하나를 짚고 갑니다. **768은 2의 제곱수가 아닙니다.** (2의 제곱수는 512, 1024, 2048입니다. 768 = 512 + 256 = 64 × 12 입니다.)
   - 실제 관례는 "2의 제곱수"가 아니라 **64의 배수**입니다. GPU가 데이터를 64개 단위로 묶어 처리할 때 가장 효율적이기 때문입니다.
   - 또한 뒤에서 배울 Multi-Head Attention에서 **헤드 수로 나누어떨어져야** 하므로, 12나 16 같은 수로 나뉘는 값이 선호됩니다. (BERT-base: 768 = 64 × 12헤드)

여기서는 `num_embeddings`는 위에서 정의한 `vocab_size(30,000)`, `embedding_dim`은 768로 정의해보겠습니다.

````python
# nn.Embedding = "정수 ID를 넣으면 벡터를 꺼내주는 거대한 표(lookup table)"
# 크기: (단어장 크기 x 벡터 차원) = (30000 x 768) = 약 2,300만 개의 학습 파라미터
embedding_vector = nn.Embedding(vocab_size, 768)
embedding_vector.weight.shape

# [주의] 첫 번째 인자는 '실제 학습된 vocab 크기'와 같거나 커야 합니다.
#        토큰 ID가 이 값을 넘으면 IndexError가 납니다.
#        안전하게 하려면 vocab_size 대신 len(tokenizer.get_vocab())를 쓰세요.
````

그러면 특정 토큰의 임베딩 벡터를 확인해보겠습니다.

````python
token_id = tokenizer.token_to_id("I")
print("token_id:", token_id)

# [방어 코드] 학습 데이터(한국어 리뷰)에 따라 "I"가 단어장에 없을 수도 있습니다.
# 그럴 경우 token_to_id는 None을 반환하고, 아래 torch.tensor에서 에러가 납니다.
if token_id is None:
    token_id = tokenizer.token_to_id("[UNK]")
    print("-> 단어장에 없어서 [UNK] 토큰으로 대체했습니다. id:", token_id)

input_id = torch.tensor([token_id], dtype=torch.long)
print("input_id 차원:", input_id.shape)   # (1,) -> 토큰 1개

vector = embedding_vector(input_id)
print("vector 차원:", vector.shape)       # (1, 768) -> 토큰 1개당 768차원 벡터
print("vector:", vector)
````

> 💡 **임베딩은 '계산'이 아니라 '조회'입니다**
>
> `nn.Embedding` 은 이름 때문에 복잡한 변환처럼 보이지만, 실제로 하는 일은 **표에서 행 하나 꺼내오기**입니다.
>
> ```
>          768차원
>        ┌──────────────────┐
> id=0   │ 0.12  -0.4  ...  │
> id=1   │ -1.3   0.8  ...  │
> id=2   │  0.5   0.1  ...  │  ← id=2를 넣으면 이 줄을 통째로 꺼내옴
>  ...   │       ...        │
> id=29999│ 0.3  -0.7  ...  │
>        └──────────────────┘
> ```
>
> **그럼 원핫 인코딩과 뭐가 다른가요?**
>
> | | 원핫 벡터 | 임베딩 벡터 |
> |---|---|---|
> | 차원 | 30,000 (단어장 크기) | 768 (우리가 정함) |
> | 값 | 하나만 1, 나머지 0 | 전부 실수 |
> | 단어 간 유사도 | **모든 쌍이 똑같이 무관** | 가까운 단어는 가까운 벡터 |
> | 학습 | 안 됨 (고정) | **학습됨** ✅ |
>
> 원핫에서는 '왕'과 '여왕'의 거리가 '왕'과 '냉장고'의 거리와 똑같습니다.
> 임베딩은 학습을 통해 **의미가 비슷한 단어를 가까이 배치**합니다.
>
> ⚠️ **지금 이 벡터는 아무 의미가 없습니다.** 방금 랜덤으로 초기화했을 뿐입니다.
> 의미는 모델을 학습시키면서 **역전파로 이 표의 값들이 조정되며** 생겨납니다.
> (`nn.Embedding` 의 weight도 학습 파라미터입니다!)


## Step 2. RNN/LSTM

### Concept Check

**Recurrent Neural Network(RNN)이란?**: 순차적(Sequential) 이전의 정보를 기억하여 현재의 정보를 처리하는 신경망 구조를 의미합니다. RNN은 다음과 같은 특징을 갖습니다.

- 입력을 순차적으로 처리합니다.
- RNN은 같은 가중치를 반복적으로 사용합니다.
- 재귀적인 구조를 가집니다.

그러면 이제부터 입력 텍스트를 RNN의 입력으로 넣어서 출력층의 결과값을 받아봅시다! 위에서 사용한 임베딩을 활용해봅시다.

````python
word_embeddings = nn.Embedding(num_embeddings=vocab_size,
                               embedding_dim=768,
                               device=device)
print("워드 임베딩 차원 :", word_embeddings.weight.shape)
````

워드 임베딩 차원에 맞게 RNN을 구현합니다.

````python
input_size: int = word_embeddings.weight.size(1) # RNN의 입력값에 들어갈 임베딩 벡터의 차원입니다. (768)
hidden_size: int = 1024  # RNN 내에서 작동하는 은닉 벡터의 차원수입니다.
num_layers: int = 1  # 쌓을 RNN layer의 개수
bidirectional: bool = False  # 단방향 RNN

rnn = nn.RNN(
    input_size=input_size,
    hidden_size=hidden_size,
    num_layers=num_layers,
    bidirectional=bidirectional,
    device=device
)

# 초기 hidden state 초기화 (num_layers * num_dirs, hidden_size)
# 문장을 읽기 전이므로 "아무것도 모르는 상태" = 0벡터에서 시작합니다.
hidden_state_shape: int = (num_layers * (2 if bidirectional else 1), hidden_size)
h_0 = torch.zeros(hidden_state_shape, device=device)
print("h_0의 차원 :", h_0.shape)

# [중요] 왜 h_0의 첫 차원이 (num_layers x 방향수) 인가?
#   층을 2개 쌓으면 각 층마다 자기 은닉 상태가 필요하고,
#   양방향이면 정방향/역방향이 각각 따로 필요하기 때문입니다.
#   지금은 1층 단방향이라 1이 됩니다.
````

> ⚠️ **PyTorch RNN의 shape 규약 — 여기서 대부분 헷갈립니다**
>
> `nn.RNN` 은 기본값이 `batch_first=False` 이고, 정식 입력은 **3차원** 입니다.
>
> ```
> 정식 (배치 있음):  (seq_len, batch_size, input_size)   ← 길이가 맨 앞!
> batch_first=True:  (batch_size, seq_len, input_size)
> ```
>
> 그런데 이번 실습은 **2차원 `(seq_len, input_size)`** 를 넣습니다. 에러가 안 나는 이유는
> PyTorch가 배치 없는(unbatched) 입력을 지원하기 때문입니다. 즉 **문장 1개짜리 특수 케이스**입니다.
>
> | 상황 | input | hidden_states | h_n |
> |---|---|---|---|
> | **지금 (배치 없음)** | (L, 768) | (L, 1024) | (1, 1024) |
> | 배치 있음 (기본) | (L, B, 768) | (L, B, 1024) | (1, B, 1024) |
> | `batch_first=True` | (B, L, 768) | (B, L, 1024) | (1, B, 1024) |
>
> 🐛 **실무 버그 1순위**: `batch_first` 를 안 맞춰서 **길이와 배치가 뒤바뀐 채로 학습**되는 경우입니다.
> 에러가 안 나고 성능만 이상해져서 찾기가 매우 어렵습니다.
> 👉 새 코드를 쓸 때는 `batch_first=True` 로 통일하는 것을 권장합니다. (`nn.Linear`, Transformer와 순서가 같아집니다)
>
> 💡 **`h_n` 은 왜 `(1, 1024)` 인가요?**
> 벡터 하나(1024차원)인데 앞에 1이 붙은 이유는 **"1층 × 1방향"** 이라는 뜻입니다.
> 2층 양방향이면 `(4, 1024)` 가 됩니다.


입력 텍스트 데이터를 위에서 배운 과정을 복습하여 임베딩 벡터로 변환합시다.

````python
text: str = "나는 학교에 간다."

# 토큰화를 진행합니다.
encoded = tokenizer.encode(text)

# 토큰의 ids만 꺼냅니다. 정수열로 텐서화를 합니다.
input_ids = encoded.ids
input_ids = torch.tensor(input_ids, dtype=torch.long, device=device)

# 변환된 정수열을 벡터화 합니다.
input_embeds = word_embeddings(input_ids)

# [확인 포인트] 세 단계의 변환을 shape으로 따라가 보세요.
print("1) 원문       :", text)
print("2) 토큰       :", encoded.tokens)
print("3) 토큰 ID    :", input_ids.tolist(), "-> shape", tuple(input_ids.shape))
print("4) 임베딩     : shape", tuple(input_embeds.shape), "= (토큰 개수, 768)")
print()
print("   문자열 -> 정수 -> 벡터. 이 3단 변환이 모든 NLP의 출발점입니다.")
````

변환된 워드 임베딩을 RNN의 입력으로 넣어 두 output을 얻습니다.

1. `hidden_states`: 각 time step에 해당하는 hidden state들의 묶음.
2. `h_n`: 모든 sequence를 거치고 나온 마지막 hidden state(`last hidden state`). hidden_states의 마지막과 동일. (`hidden_states[-1]`과 동일)

````python
outputs = rnn(input_embeds, h_0) # hidden_states, h_n
hidden_states = outputs[0] # (sequence_length, hidden_size)
h_n = outputs[1]

# sequence_length (L) : input_token의 길이 (5)
# embedding_dim (H_in): 입력값의 벡터 차원 수 (768)
# hidden_size (H_out) : hidden state 차원 수 (1024)
# num_dirs (D)        : 방향의 개수 (1)
# num_layers          : layer 개수 (1)
print("워드 임베딩 차원 : ", input_embeds.shape)  # (L, H_in)
print("hidden_states 차원 : ", hidden_states.shape)  # (L, D * H_out)
print("h_n 차원 : ", h_n.shape)  # (D * num_layers, H_out))

if torch.equal(hidden_states[-1].unsqueeze(0), h_n):
    print("hidden_states의 마지막과 h_n이 같습니다.")
````

> 💡 **`hidden_states` 와 `h_n`, 뭐가 다른가요?**
>
> RNN은 토큰을 **하나씩 순서대로** 읽으면서, 읽을 때마다 은닉 상태를 갱신합니다.
>
> ```
> [CLS] -> 나는 -> 학교 -> ##에 -> 간다 -> . -> [SEP]
>   h1     h2      h3      h4      h5    h6    h7
>   └────────── hidden_states (전부 모은 것) ──────────┘
>                                              └─ h_n (마지막 것)
> ```
>
> | | 내용 | 쓰임 |
> |---|---|---|
> | `hidden_states` | **매 시점**의 은닉 상태 전부 | Attention (Step 3에서 사용!) |
> | `h_n` | **마지막** 시점의 은닉 상태 | 문맥 벡터(context vector) |
>
> ⚠️ **"`h_n` == `hidden_states[-1]`" 은 조건부로만 참입니다**
> 지금처럼 **1층 · 단방향**일 때만 같습니다.
> - `num_layers=2` 라면: `h_n` 은 **모든 층**의 마지막 상태를 담고, `hidden_states` 는 **맨 위 층**의 것만 담습니다.
> - `bidirectional=True` 라면: `hidden_states[-1]` 은 정방향 마지막 + 역방향 첫 번째가 섞인 값이라 아예 다릅니다.
>
> 👉 위 셀의 `torch.equal(...)` 검사가 통과했다면, "지금 설정에서는 같다"는 뜻으로만 받아들이세요.


결과적으로 RNN은 입력 벡터를 받아 또다른 은닉 상태 (hidden state)를 얻습니다. 그러면 이러한 은닉 상태를 얻어서 어떠한 작업을 할 수 있을까요?

### Concept Check

**은닉 상태(hidden state)는 문장의 정보들을 압축적으로 저장합니다.**: RNN layer를 통과하면서 문장 전체의 정보를 압축하게 되고 이러한 정보들은 hidden state에 담기게 됩니다. 이러한 hidden state는 문맥 벡터(context vector)로 사용됩니다. 문맥 벡터(context vector)는 입력 문장의 정보들을 벡터상에 압축하여 저장한 것으로, 이를 통해 다양한 task를 수행할 수 있게 됩니다.

여기서는 번역(translation) task를 수행하기 위해 hidden state를 사용하겠습니다. 번역을 하기 위해서는 last hidden state를 다시 저희의 입력 데이터와 유사한 형태인 텍스트(토큰) id로 변환하는 layer가 필요합니다. 이를 저희는 Decoder라고 부릅니다.

그러면 아래에서 Encoder와 Decoder를 연결하여 번역을 수행하는 모델을 구현하겠습니다. 먼저 인코더를 구현하겠습니다. 위에서 구현한 `torch.nn.rnn`을 그대로 이용하여 클래스화를 진행하는 것과 동일합니다.

### TODO 1: RNN Encoder 구현

**요구사항**:
1. `__init__`: RNN에 입력을 넣어 은닉 상태를 얻기 위해 필요한 두 가지 요소를 속성으로 선언합니다.
   1. `self.word_embedding`: 인자로 받은 `vocab_size`, `embedding_dim`을 활용해 토큰화된 결과를 임베딩할 수 있도록 `nn.Embedding`을 선언해줍니다.
   2. `self.rnn`: 인자로 받은 `hidden_size`, `num_layers`, `bidirectional`을 통해 RNN을 선언해줍니다.
2. `forward`: 정수열로 인코딩된 `input_ids`를 받았을 때 RNN을 통과한 은닉상태 임베딩을 반환합니다.
   1. 입력 받은 정수열을 벡터화합니다.
   2. 1에서 벡터화한 입력을 RNN에 통과시켜 은닉상태를 얻어줍니다.

````python
from abc import ABC, abstractmethod

# 인코더 모델은 RNN을 사용합니다. 아래는 추상화 클래스입니다.
class Encoder(nn.Module, ABC):
    def __init__(self: "Encoder") -> None:
        super().__init__()
        pass

    @abstractmethod
    def forward(self: "Encoder", input_ids: torch.Tensor) -> torch.Tensor:
        pass


class RNNEncoder(Encoder):
    def __init__(
        self,
        vocab_size: int,
        embedding_dim: int,
        hidden_size: int,
        num_layers: int,
        bidirectional: bool,
    ) -> None:
        super().__init__()
        self.word_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.rnn = nn.RNN(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            num_layers=num_layers,
            bidirectional=bidirectional,
        )

    def forward(self, input_ids):
        # 입력 토큰을 워드 임베딩을 통해 임베딩 변환을 합니다.
        input_embeds = self.word_embeddings(input_ids)

        # RNN을 통해 입력 임베딩을 문맥 벡터(context vector)화 합니다.
        hidden_states, h_n = self.rnn(input_embeds)
        return hidden_states, h_n

vocab_size = 30000
embedding_dim = 768
hidden_size = 1024  # RNN의 hidden size
num_layers = 1  # 쌓을 RNN layer의 개수
bidirectional = False  # 단방향 RNN

rnn_encoder = RNNEncoder(
    vocab_size=vocab_size,
    embedding_dim=embedding_dim,
    hidden_size=hidden_size,
    num_layers=num_layers,
    bidirectional=bidirectional
).to(device)

outputs = rnn_encoder(input_ids)
hidden_states = outputs[0]
h_n = outputs[1]
print("hidden_states 차원 : ", hidden_states.shape)  # (L, H_out)
print("h_n 차원 : ", h_n.shape)  # (num_layers*D, H_out) = (1, H_out)
````

### TODO 2: RNN Decoder 구현

이번에 디코더 부분을 구현하겠습니다. 디코더 또한 RNN을 사용하지만 인코더보다 한 가지의 추가 과정이 필요합니다. 바로 매 step의 hidden state를 `nn.Linear` 레이어를 통과해 기존의 `vocab_size`의 단어장(토큰) 중 하나의 토큰을 출력하도록 확률을 예측해야합니다.

**요구사항**: `__init__`은 `RNNEncoder`와 유사하여 이번에는 `forward`만 구현해봅시다. 주어진 `max_len`까지 한 스텝씩 다음에 나올 단어를 예측하는 과제를 수행해주세요.
1. RNNEncoder의 마지막 은닉상태인 `h_n`을 입력으로 받아 한 step씩 은닉 벡터를 만들어줍니다.
2. 위에서 만든 은닉벡터를 `nn.Linear`를 거쳐 토큰을 예측하여 `logit`을 만들어줍니다.
3. `logit`을 기반으로 예측된 토큰인 `output_token`을 `argmax`를 통해 찾아줍니다
4. 만약 출력값이 문장의 종료를 나타내는 [SEP]이 나오면 `break`를 통해 생성을 멈추고, 아닐 경우 `input_token`에 `output_token`을 할당하여 루프를 계속 진행시킵니다.

````python
class Decoder(nn.Module, ABC):
    def __init__(self: "Decoder"):
        super().__init__()

    @abstractmethod
    def forward(self,
                input_ids: torch.Tensor,
                init_hidden_state: torch.Tensor) -> torch.Tensor:
        pass


class RNNDecoder(Decoder):
    def __init__(
        self,
        vocab_size: int,
        embedding_dim: int,
        hidden_size: int,
        num_layers: int,
        bidirectional: bool,
        start_token_id: int,
        end_token_id: int,
    ):
        super().__init__()
        self.start_token_id = start_token_id
        self.register_buffer("input_token", torch.tensor([self.start_token_id], dtype=torch.long))
        self.end_token_id = end_token_id

        self.word_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.rnn = nn.RNN(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            num_layers=num_layers,
            bidirectional=bidirectional,
        )
        # hidden_size -> vocab_size: "이 은닉 상태 다음에 올 단어는?"을 3만 개 후보 점수로 변환
        self.fully_connected_layer = nn.Linear(hidden_size, vocab_size)
        # ⚠️ 아래 한 줄은 주의가 필요합니다.
        #    __init__ 시점(=.to(device) 이전)의 device가 그대로 박제됩니다.
        #    나중에 모델을 GPU로 옮겨도 self.device는 여전히 'cpu'입니다.
        #    지금 코드에서는 사용하지 않아 문제가 없지만, 실무에서 이런 패턴은 버그의 씨앗입니다.
        #    필요하다면 사용하는 시점에 next(self.parameters()).device 로 조회하세요.
        self.device = self.word_embeddings.weight.device

    def forward(self, init_hidden_state, max_len: int = 10):
        logits = []
        input_token = self.input_token
        output_token_ids = [input_token.item()] # tensor에서 item()을 사용하여 int로 변환합니다.
        h_n = init_hidden_state # h_n은 encoder의 h_0와 동일한 역할을 합니다. Encoder에서 입력된 문장 전체의 은닉 상태를 나타냅니다.

        # 주어진 `max_len`까지 한 스텝씩 다음에 나올 단어를 예측하는 과제를 수행해주세요.
        for _ in range(max_len):
            # 1. RNNEncoder의 마지막 `h_n`을 입력으로 받아 한 step씩 은닉 벡터를 만들어줍니다.
            embedded = self.word_embeddings(input_token)  # 직전 입력 토큰만 사용 (1, H_in)
            _, h_n = self.rnn(embedded, h_n) # 마지막 state만 사용 (1, H_out)

            # 2. 위에서 만든 은닉벡터를 `nn.Linear`를 거쳐 토큰을 예측하여 `logit`을 만들어줍니다.
            logit = self.fully_connected_layer(h_n) # (1, vocab_size)
            logit = logit.squeeze()
            logits.append(logit)

            # 3. `logit`을 기반으로 예측된 토큰인 `output_token`을 `argmax`를 통해 찾아줍니다.
            output_token = torch.argmax(logit, dim=-1).unsqueeze(0)
            output_token_ids.append(output_token.item())

            # 4. 만약 출력값이 문장의 종료를 나타내는 [SEP]이 나오면 `break`를 통해 생성을 멈추고,
            # 아닐 경우 `input_token`에 `output_token`을 할당하여 루프를 계속 진행시킵니다.
            if output_token == self.end_token_id:
                # 문장의 종료를 의미하는 special token([SEP])이 나왔다면 추론(생성)을 종료합니다.
                break
            else:
                input_token = output_token

        logits = torch.stack(logits, dim=0)  # (max_len, vocab_size)
        return logits, output_token_ids

start_token_id: int = tokenizer.encode("[CLS]").ids[0]
end_token_id: int = tokenizer.encode("[SEP]").ids[0]

vocab_size: int = 30000
embedding_dim: int = 768
hidden_size: int = 1024  # RNN의 hidden size
num_layers: int = 1  # 쌓을 RNN layer의 개수
bidirectional: bool = False  # 단방향 RNN

rnn_decoder = RNNDecoder(
    vocab_size=vocab_size,
    embedding_dim=embedding_dim,
    hidden_size=hidden_size,
    num_layers=num_layers,
    bidirectional=bidirectional,
    start_token_id=start_token_id,
    end_token_id=end_token_id,
).to(device)
logits, output_token_ids = rnn_decoder(h_n)
output_texts = tokenizer.decode(output_token_ids)
print(output_texts)
````

> 💡 **이 디코더가 쓰는 생성 방식 — Greedy Decoding**
>
> 매 스텝 `torch.argmax(logit)` 으로 **가장 점수가 높은 토큰 하나**만 고르고 있습니다.
> 이를 **탐욕적 디코딩(greedy decoding)** 이라고 합니다. 단순하지만 약점이 있습니다.
>
> | 방식 | 방법 | 특징 |
> |---|---|---|
> | **Greedy** | 매번 1등만 선택 | 빠름. **같은 말 반복**에 빠지기 쉬움 |
> | **Beam Search** | 후보 k개를 동시에 추적 | 번역에 강함. 느림 |
> | **Sampling (top-k / top-p)** | 확률에 따라 뽑기 | 다양하고 창의적. ChatGPT가 사용 |
>
> 👉 `temperature`, `top_p` 같은 LLM 옵션이 바로 이 **샘플링 방식을 조절하는 손잡이**입니다.
> Step 4에서 쓸 `model.generate()` 안에 이 모든 전략이 들어 있습니다.

> ⚠️ **학습 코드가 아니라 '추론 코드'입니다 — Teacher Forcing 이야기**
>
> 지금 디코더는 **자기가 뱉은 토큰을 다시 자기 입력으로** 넣습니다. (`input_token = output_token`)
> 그런데 **학습할 때는 이렇게 하지 않습니다.**
>
> ```
> 추론 시: 모델이 뱉은 것을 다시 입력   -> 한 번 틀리면 뒤가 전부 무너짐
> 학습 시: 정답 문장을 그대로 입력       -> 이것을 Teacher Forcing 이라고 부릅니다
> ```
>
> 학습을 이 코드처럼 하면, 초반에 엉터리 출력이 계속 입력으로 들어가서 학습이 거의 되지 않습니다.
> 👉 학습/추론에서 디코더가 다르게 동작한다는 점, 나중에 Seq2Seq를 직접 학습시킬 때 반드시 기억하세요.


이제 구현한 encoder와 decoder를 연결하여 seq2seq 모델을 구현해보겠습니다.

````python
class RNNSeq2Seq(nn.Module):
    def __init__(self: "RNNSeq2Seq", encoder: nn.Module, decoder: nn.Module) -> None:
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self: "RNNSeq2Seq", input_ids):
        _, context_vector = self.encoder(input_ids) # encoder에서 생성한 context_vector(h_n)을 decoder layer로 전달
        logits, output_tokens = self.decoder(context_vector)
        return logits, output_tokens

seq2seq = RNNSeq2Seq(rnn_encoder, rnn_decoder)
logits, output_tokens = seq2seq(input_ids)
output_token_ids = logits.argmax(dim=-1)
output_texts = tokenizer.decode(output_token_ids.tolist())
print(output_texts)
````

RNN뿐만 아니라 LSTM, 어텐션 등을 사용하여 Seq2Seq 구조를 구현할 수 있습니다. 전체적인 큰 틀은 그대로 유지한 채, RNN 모듈만 바꿔주기만 하면 됩니다. 그러면 이제부터 LSTM으로 다시 한번 구현해보겠습니다.

### Concept Check
**LSTM(Long-Short term Memory)**: RNN을 직접 구현해보셔서 아시겠지만, 시퀀스가 길어질수록 앞의 hidden state의 내용이 점점 사라질 가능성이 높습니다. 이를 대처하기 위해 LSTM에는 장기 기억을 담당하는 cell state를 통해 좀 더 성능을 높였습니다.

그러면 Encoder에서 LSTM을 적용해보겠습니다. 기본구조는 RNNEncoder와 거의 유사합니다. 중요한 점은 RNN에는 없던 [**cell_state(`c_n`)** 이 추가되었다는](https://docs.pytorch.org/docs/stable/generated/torch.nn.LSTM.html) 점입니다. 아래 코드를 참고해주세요.

````python
class LSTMEncoder(Encoder):
    def __init__(
        self,
        vocab_size: int,
        embedding_dim: int,
        hidden_size: int,
        num_layers: int,
        bidirectional: bool,
    ):
        super().__init__()
        self.word_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            num_layers=num_layers,
            bidirectional=bidirectional,
        )

    def forward(self, input_ids):
        input_embeds = self.word_embeddings(input_ids)
        hidden_states, (h_n, c_n) = self.lstm(input_embeds)
        return hidden_states, (h_n, c_n)

vocab_size = 30000
embedding_dim = 768
hidden_size = 1024  # RNN의 hidden size
num_layers = 1  # 쌓을 RNN layer의 개수
bidirectional = False  # 단방향 RNN

lstm_encoder = LSTMEncoder(
    vocab_size=vocab_size,
    embedding_dim=embedding_dim,
    hidden_size=hidden_size,
    num_layers=num_layers,
    bidirectional=bidirectional
).to(device)

hidden_states, (h_n, c_n) = lstm_encoder(input_ids)
print("hidden_states 차원 : ", hidden_states.shape)  # (L, B, d_h)
print("h_n 차원 : ", h_n.shape)  # (num_layers*num_dirs, B, d_h) = (1, d_h)
print("c_n 차원 : ", c_n.shape)  # (num_layers*num_dirs, B, d_h) = (1, d_h)
````

> 💡 **RNN과 LSTM, 무엇이 달라졌나 — "메모장을 하나 더 들고 다닌다"**
>
> RNN은 은닉 상태 `h` 하나로 모든 걸 기억해야 합니다. 새 정보가 들어올 때마다 통째로 덮어써야 하니,
> 문장이 길어지면 **앞부분이 지워집니다.** (장기 의존성 문제)
>
> LSTM은 여기에 **셀 상태 `c`** 라는 별도의 메모장을 추가했습니다.
>
> | | 역할 | 비유 |
> |---|---|---|
> | `h` (hidden state) | 지금 당장 쓸 정보 | **단기 기억** — 작업 중인 메모 |
> | `c` (cell state) | 오래 유지할 정보 | **장기 기억** — 노트에 적어둔 내용 |
>
> 그리고 3개의 **게이트**가 이 메모장을 관리합니다.
>
> | 게이트 | 하는 일 |
> |---|---|
> | Forget gate | 노트에서 **뭘 지울지** 결정 |
> | Input gate | 새 정보 중 **뭘 적을지** 결정 |
> | Output gate | 노트에서 **뭘 꺼내 쓸지** 결정 |
>
> ⚠️ **하지만 LSTM도 근본적인 한계가 있습니다.**
> 결국 **문장 전체를 벡터 하나(`h_n`)에 욱여넣어야** 한다는 점은 그대로입니다.
> 100단어짜리 문장도, 5단어짜리 문장도 똑같이 1024차원 벡터 하나로 압축됩니다.
>
> 👉 이 병목을 뚫으려는 시도가 바로 다음 Step에서 배울 **Attention** 입니다.


이번에는 LSTM을 사용하여 `LSTMDecoder`를 구현해보겠습니다. `RNNDecoder`와 로직이 같습니다.

````python
class LSTMDecoder(Decoder):
    def __init__(
        self,
        vocab_size: int,
        embedding_dim: int,
        hidden_size: int,
        num_layers: int,
        bidirectional: bool,
        start_token_id: int,
        end_token_id: int,
    ):
        super().__init__()
        self.start_token_id = start_token_id
        self.register_buffer("input_token", torch.tensor([self.start_token_id], dtype=torch.long))
        self.end_token_id = end_token_id

        self.word_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            num_layers=num_layers,
            bidirectional=bidirectional,
        )
        self.fully_connected_layer = nn.Linear(hidden_size, vocab_size)

    def forward(self, init_hidden_state, init_cell_state, max_len: int = 10):
        logits = []
        input_token = self.input_token
        output_token_ids = [input_token.item()] # tensor에서 item()을 사용하여 int로 변환합니다.
        h_n = init_hidden_state # h_n은 encoder의 h_0와 동일한 역할을 합니다.
        c_n = init_cell_state

        for _ in range(max_len):
            embedded = self.word_embeddings(input_token) # 직전 입력 토큰을 사용 [1, embedding_dim]
            _, (h_n, c_n) = self.lstm(embedded, (h_n, c_n))

            logit = self.fully_connected_layer(h_n)
            logit = logit.squeeze()
            logits.append(logit)

            output_token = torch.argmax(logit, dim=-1).unsqueeze(0)
            output_token_ids.append(output_token.item())

            if output_token == self.end_token_id:
                # 문장의 종료를 의미하는 special token([SEP])이 나왔다면 추론(생성)을 종료합니다.
                break
            else:
                input_token = output_token

        logits = torch.stack(logits, dim=0) # (max_len, vocab_size)
        return logits, output_token_ids


start_token_id: int = tokenizer.encode("[CLS]").ids[0]
end_token_id: int = tokenizer.encode("[SEP]").ids[0]

vocab_size: int = 30000
embedding_dim: int = 768
hidden_size: int = 1024 # RNN의 hidden size
num_layers: int = 1 # 쌓을 RNN layer의 개수
bidirectional: bool = False # 단방향 RNN

lstm_decoder = LSTMDecoder(
    vocab_size=vocab_size,
    embedding_dim=embedding_dim,
    hidden_size=hidden_size,
    num_layers=num_layers,
    bidirectional=bidirectional,
    start_token_id=start_token_id,
    end_token_id=end_token_id,
).to(device)

logits, output_tokens = lstm_decoder(h_n, c_n)
output_token_ids = logits.argmax(dim=-1)
output_texts = tokenizer.decode(output_token_ids.tolist())
print(output_texts)
````

Encoder와 Decoder를 사용하여 Seq2Seq 모델을 구현해보겠습니다.

````python
class LSTMSeq2Seq(nn.Module):
    def __init__(self, encoder: nn.Module, decoder: nn.Module):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, input_ids):
        # encoder에서 생성한 context_vector(h_n)을 decoder layer로 전달
        _, (context_vector, cell_states) = self.encoder(input_ids)
        logits, output_tokens = self.decoder(context_vector, cell_states)
        return logits, output_tokens

seq2seq = LSTMSeq2Seq(lstm_encoder, lstm_decoder)
logits, output_tokens = seq2seq(input_ids)
output_token_ids = logits.argmax(dim=-1)
output_texts = tokenizer.decode(output_token_ids.tolist())
print(output_texts)
````

## Step 3. Attention Mechanism

### Concept Check

**Attention Mechanism**: 현재 구현할 seq2seq 모델에서의 Attention은 최근 사용하는 attention은 아닙니다. 최근의 Transformers 모델들은 Multi-Head Scaled Dot-Product Attention을 사용합니다. 해당 내용은 과제에서 다룰 예정입니다.

1. 전체적인 Seq2Seq 모델의 구조는 동일합니다.
2. Encoder에서 context vector를 얻을 때, LSTM을 사용하는 Encoder 모듈을 그대로 사용합니다.
3. Decoder에서 output token을 생성할 때, attention mechanism을 추가합니다.

### TODO 3: Luong Attention 구현

이제 Luong Attention을 구현합니다. 어텐션의 주된 목표는 현재 단계의 `h_t`이 전체 시퀀스의 토큰과 어떤 관계를 가지는지 보는 것입니다. 그래서 다음과 같은 스텝을 거칩니다.
1. `h_t`을 `nn.Linear`를 통해 같은 형태이지만 새롭게 임베딩한 `Wa_ht`를 구합니다.
2. `Wa_ht`와 전체 시퀀스 사이의 관계를 보기 위해 둘을 내적합니다
3. 내적한 결과에 `softmax` 연산을 걸어 확률화합니다. 이는 현재 출력된 은닉상태 `h_t`가 전체 시퀀스와 어떤 관계를 가지는지 나타냅니다.
4. 3의 확률화 결과를 통해 전체 시퀀스의 은닉상태를 가중합하여 `context_vector`를 만들어줍니다.

여기서 얻어낸 `context_vector`와 기존 `h_t`를 결합하여 다음 토큰을 출력하면 기존 시퀀스의 정보와의 관계를 보며 출력을 할 수 있어 더 높은 성능을 기대할 수 있습니다.

📌 **점수 계산 방식(score function)에 대한 정확한 명칭**

Luong 논문(2015)은 attention score를 구하는 방법을 3가지로 제시했습니다. 아래 코드는 이 중 **general** 방식입니다.

| 방식 | 수식 | 학습 파라미터 |
|---|---|---|
| **dot** | $h_t^\top \bar{h}_s$ | 없음 |
| **general** ← 이번 구현 | $h_t^\top W_a \bar{h}_s$ | $W_a$ 있음 |
| **concat** | $v_a^\top \tanh(W_a[h_t; \bar{h}_s])$ | $W_a, v_a$ 있음 |

즉 `self.W_a = nn.Linear(...)` 가 있기 때문에 순수한 dot이 아니라 **general** 입니다.
(파라미터가 하나 더 있어서 "어떤 방향으로 비교할지"까지 학습할 수 있습니다.)

**요구사항**: 위에서 제시한 4단계를 코드 안에 수식을 포함하여 구체적으로 작성하였습니다. 지시에 따라 Luong Attention 구현을 진행해주세요.

````python
class LuongAttention(nn.Module):
    """Luong(2015)의 general score 방식 어텐션.

    Args:
        hidden_size (int): 인코더/디코더 은닉 상태의 차원

    Note:
        forward에 @torch.no_grad()가 붙어 있습니다. 지금은 학습 없이 구조만
        확인하는 단계라 괜찮지만, 실제로 학습시킬 때는 반드시 제거해야 합니다.
        붙어 있으면 W_a에 기울기가 흐르지 않아 어텐션이 전혀 학습되지 않습니다.
    """

    def __init__(self, hidden_size: int):
        super().__init__()
        # bias=False인 이유: 순수한 '방향 변환'만 학습하면 되기 때문입니다.
        self.W_a = nn.Linear(hidden_size, hidden_size, bias=False)

    @torch.no_grad()  # ⚠️ 학습 시 반드시 제거! (제거하지 않으면 W_a가 학습되지 않습니다)
    def forward(self, h_t, encoder_outputs):
        """현재 디코더 은닉 상태와 인코더 전체 출력의 관계를 계산한다.

        Args:
            h_t (torch.Tensor): 디코더의 현재 은닉 상태. (1, H)
            encoder_outputs (torch.Tensor): 인코더의 모든 시점 은닉 상태. (L, H)

        Returns:
            tuple: (context_vector (1, H), attention_weights (L, 1))
        """
        # 1. hidden state를 W_a에 projection하여 Wa_ht를 구합니다.
        # Wa_ht = W_a * h_t (1, H_out)
        # -> "인코더 출력과 비교하기 좋은 형태"로 한 번 변환하는 단계입니다.
        Wa_ht = self.W_a(h_t)

        # 2. encoder_outputs와 Wa_ht를 내적하여 attention score를 구합니다.
        # score = encoder_outputs * (Wa_ht)^T
        # (L, H_out) @ (H_out, 1) -> (L, 1)
        # -> 내적이 크다 = 방향이 비슷하다 = 지금 필요한 정보가 거기 있다
        attention_score = encoder_outputs @ Wa_ht.T

        # 3. attention score를 softmax layer에 통과시켜 attention weights(attention distribution)을 구합니다.
        # -> 점수를 '합이 1인 확률'로 바꿔 "어디를 몇 % 볼지"를 정합니다.
        attention_weights = F.softmax(attention_score, dim=-1)

        # 4. 각 encoder의 attention weights와 encoder의 hidden state를 내적하여 context vector(attention value)를 구합니다.
        # context = weights^T * encoder_outputs
        # (1, Seq_Len) @ (Seq_Len, Hidden) -> (1, Hidden)
        # -> 확률을 가중치로 삼아 인코더 상태들을 '섞어' 하나의 벡터로 만듭니다.
        context_vector = attention_weights.T @ encoder_outputs
        return context_vector, attention_weights
````

> 💡 **Attention을 한 문장으로 — "시험 볼 때 교과서를 다시 펼쳐보는 것"**
>
> | | 방식 | 한계 / 장점 |
> |---|---|---|
> | **Seq2Seq (Attention 없음)** | 교과서를 다 읽고 **요약 한 장**만 들고 시험장에 들어감 | 긴 내용은 요약에서 누락 |
> | **Attention** | **교과서를 통째로** 들고 가서, 문제마다 필요한 페이지를 펼쳐봄 | 정보 손실 없음 ✅ |
>
> 코드로 보면 딱 이 차이입니다.
> ```
> Attention 없음: decoder(h_n)                    ← 마지막 상태 하나만
> Attention 있음: decoder(h_n, encoder_outputs)   ← 전부 다 넘겨줌
> ```
>
> 🔑 **`attention_weights` 는 그 자체로 값진 산출물입니다**
>
> 이 값은 "지금 이 단어를 만들 때 입력의 어디를 봤는가"를 알려줍니다.
> 번역 모델에서 이걸 히트맵으로 그리면 **단어 정렬(alignment)** 이 눈에 보입니다.
>
> ```
>          나는   학교   ##에   간다
>   I      0.71  0.09   0.08  0.12    ← "I"를 만들 때 "나는"을 71% 봄
>   go     0.10  0.12   0.11  0.67
>   to     0.08  0.31   0.49  0.12
>   school 0.11  0.68   0.13  0.08
> ```
>
> 👉 **딥러닝의 블랙박스를 들여다보는 몇 안 되는 창**입니다.
> 실무에서 모델이 왜 그런 답을 냈는지 설명할 때 자주 활용합니다.
> (지금은 학습 전이라 균등하게 나올 것입니다. `_` 로 버리지 말고 한번 출력해보세요!)


잘 구현되었는지 확인해봅시다. 다음 코드는 여러분들이 구현한 Attention 모듈을 활용한 Seq2Seq 모듈입니다. 주석을 참고하며 어떤 부분이 추가되었는지 확인해보세요.

````python
class AttentionDecoder(nn.Module):
    def __init__(
        self: "AttentionDecoder",
        vocab_size: int,
        embedding_dim: int,
        hidden_size: int,
        num_layers: int,
        bidirectional: bool,
        start_token_id: int,
        end_token_id: int,
    ):
        super().__init__()
        self.start_token_id = start_token_id
        self.register_buffer("input_token", torch.tensor([self.start_token_id], dtype=torch.long))
        self.end_token_id = end_token_id
        self.word_embeddings = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(
            input_size=embedding_dim,
            hidden_size=hidden_size,
            num_layers=num_layers,
            bidirectional=bidirectional,
        )

        # Attention Mechanism을 위한 레이어
        self.attn = LuongAttention(hidden_size)
        self.W_c = nn.Linear(hidden_size * 2, hidden_size)

        # 토큰 출력을 위한 레이어
        self.fully_connected_layer = nn.Linear(hidden_size, vocab_size)

    @torch.no_grad()  # 학습 시 제거
    def forward(
        self,
        init_hidden_state,
        init_cell_state,
        encoder_outputs,
        max_len: int = 10,
    ):
        logits = []
        input_token = self.input_token
        output_token_ids = [input_token.item()] # tensor에서 item()을 사용하여 int로 변환합니다.
        h_n = init_hidden_state # h_n은 encoder의 h_0와 동일한 역할을 합니다.
        c_n = init_cell_state

        for _ in range(max_len):
            # LSTMEncoder 마지막 `h_n`을 입력으로 받아 한 step씩 은닉 벡터를 만들어줍니다.
            embedded = self.word_embeddings(input_token)  # 직전 입력 토큰만 사용 [1, embedding_dim]
            _, (h_n, c_n) = self.lstm(embedded, (h_n, c_n))   # outputs: [S,B,D*H] or [B,S,D*H]

            # 어텐션을 계산하여 `context_vector`를 만들어줍니다.
            # `h_n`과 `context_vector`를 연결하여 v_t를 만들고
            # trainable weights에 통과시키고 tanh를 적용합니다.
            context_vector, _ = self.attn(h_n, encoder_outputs)
            v_t = torch.cat([h_n, context_vector], dim=-1)
            attentional_hidden_state = torch.tanh(self.W_c(v_t))

            # 위에서 만든 은닉벡터를 `nn.Linear`를 거쳐 토큰을 예측하여 `logit`을 만들어줍니다.
            logit = self.fully_connected_layer(attentional_hidden_state)
            logit = logit.squeeze()
            logits.append(logit)

            # `logit`을 기반으로 예측된 토큰인 `output_token`을 `argmax`를 통해 찾아줍니다.
            output_token = torch.argmax(logit, dim=-1).unsqueeze(0)
            output_token_ids.append(output_token.item())

            # 만약 출력값이 문장의 종료를 나타내는 [SEP]이 나오면 `break`를 통해 생성을 멈추고,
            # 아닐 경우 `input_token`에 `output_token`을 할당하여 루프를 계속 진행시킵니다.
            if output_token == self.end_token_id:
                # 문장의 종료를 의미하는 special token([SEP])이 나왔다면 추론(생성)을 종료합니다.
                break
            else:
                input_token = output_token

        logits = torch.stack(logits, dim=0) if logits else torch.empty(0, self.fully_connected_layer.out_features)
        return logits, output_token_ids

start_token_id: int = tokenizer.encode("[CLS]").ids[0]
end_token_id: int = tokenizer.encode("[SEP]").ids[0]

vocab_size: int = 30000
embedding_dim: int = 768
hidden_size: int = 1024 # RNN의 hidden size
num_layers: int = 1 # 쌓을 RNN layer의 개수
bidirectional: bool = False # 단방향 RNN

attention_decoder = AttentionDecoder(
    vocab_size=vocab_size,
    embedding_dim=embedding_dim,
    hidden_size=hidden_size,
    num_layers=num_layers,
    bidirectional=bidirectional,
    start_token_id=start_token_id,
    end_token_id=end_token_id,
).to(device)

logits, output_tokens = attention_decoder(h_n, c_n, hidden_states)
output_token_ids = logits.argmax(dim=-1)
output_texts = tokenizer.decode(output_token_ids.tolist())
print(output_texts)
````

Decoder layer를 구현했으니 이제 Seq2Seq 모델에 적용해봅니다.

````python
class AttentionSeq2Seq(nn.Module):
    def __init__(self, encoder: nn.Module, decoder: nn.Module) -> None:
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, input_ids):
        hidden_states, (last_hidden_state, cell_states) = self.encoder(input_ids) # encoder에서 생성한 h_n을 decoder layer로 전달
        logits, output_tokens = self.decoder(last_hidden_state, cell_states, hidden_states)
        return logits, output_tokens

seq2seq = AttentionSeq2Seq(lstm_encoder, attention_decoder)
logits, output_tokens = seq2seq(input_ids)
output_token_ids = logits.argmax(dim=-1)
output_texts = tokenizer.decode(output_token_ids.tolist())
print(output_texts)
````

이미 학습된 모델을 통해 추론을 진행합니다.
위의 실습에서 추론했던 것과는 다르게 학습된 모델이므로 성능이 더 높게 나타납니다.

````python
text = "나는 학교에 간다."
"""여기서는 batch로 입력을 처리하여 차원이 [seq_len]이 아닌 [batch_size, seq_len]입니다. 여기서는 입력이 한개이므로 [1, seq_len]입니다."""
# 직접 만든 토크나이저와 달리, HF 토크나이저는 호출만 하면 모델이 바로 먹을 수 있는 dict를 줍니다.
# return_tensors="pt" -> PyTorch 텐서로 반환 (input_ids, attention_mask 포함)
encoded = tokenizer(text, return_tensors="pt")
print("입력 키:", list(encoded.keys()))
print("input_ids shape:", encoded['input_ids'].shape)   # (1, seq_len)

# model.generate()는 우리가 Step 2에서 손으로 짠 while 루프를 대신해줍니다.
#   - 다음 토큰 예측 -> 이어붙이기 -> EOS 나오면 종료
#   - beam search, sampling, repetition penalty 등이 전부 내장되어 있습니다.
generated_ids = model.generate(
    **encoded,
    max_new_tokens=64,   # 최대 64개까지만 생성 (무한 루프 방지)
)

# skip_special_tokens=True -> </s>, <pad> 같은 특수 토큰을 결과에서 제거
translation = tokenizer.decode(generated_ids.squeeze(), skip_special_tokens=True)
print("SRC:", text)
print("MT :", translation)

# [비교 포인트] Step 2에서 우리가 만든 모델의 출력과 비교해보세요.
#   같은 Encoder-Decoder 구조인데 결과가 천지차이입니다.
#   차이는 오직 하나 - '학습되었는가'입니다. 구조가 전부가 아닙니다.
````

## Step 5. 아키텍처별 모델 다뤄보기(Encoder model, Decoder model)
지금까지는 Seq2Seq(Encoder - Decoder) 모델 구조를 다뤘습니다. 하지만, 현재 가장 많이 사용되는 모델은 Decoder-only 모델입니다.

| 구조 | 대표 모델 | 학습 방식 | 잘하는 일 | 현재 위상 |
|---|---|---|---|---|
| **Encoder-only** | BERT, RoBERTa | 빈칸 맞추기 (MLM) | 문장 **이해**, 분류, 검색 | 임베딩·검색(RAG)에서 여전히 현역 |
| **Decoder-only** | GPT, LLaMA, Claude | 다음 단어 맞추기 | 문장 **생성**, 대화 | **주류** |
| **Encoder-Decoder** | T5, BART, Whisper | 입력 -> 출력 변환 | 번역, 요약, 음성인식 | 특정 분야에서 **현역** |

📌 **"Encoder-Decoder는 이제 안 쓴다"는 오해입니다.**
범용 챗봇 영역에서 Decoder-only가 주류가 된 것은 맞지만, **입력과 출력이 뚜렷하게 구분되는 작업**에서는
Encoder-Decoder가 여전히 표준입니다.
- 음성 인식: **Whisper** (OpenAI)
- 기계 번역: **NLLB** (Meta), 방금 써본 Helsinki-NLP 모델
- 요약·구조화 변환: **T5**, **BART**

핵심은 "무엇이 최신인가"가 아니라 **"내 문제에 어떤 구조가 맞는가"** 입니다.

그러면 Only Encoder 모델과 Only Decoder 모델을 이용해 모델 추론을 해보겠습니다.

> 💡 **왜 BERT는 문장을 못 만들고, GPT는 문맥을 덜 보나요? — 마스킹의 차이**
>
> 두 모델의 차이는 **"각 토큰이 어디까지 볼 수 있는가"** 하나입니다.
>
> ```
> BERT (양방향)                    GPT (단방향 = causal)
> "나는 [MASK] 간다"                "나는 학교에 간다"
>       ↑ ↑ ↑                        →  →  →
>   앞뒤를 모두 봄                   앞만 볼 수 있음
>
> 장점: 문맥 이해가 깊음            장점: 다음 단어를 생성할 수 있음
> 단점: 이어서 쓸 수 없음           단점: 뒤 문맥을 못 봄
> ```
>
> BERT가 문장을 생성하지 못하는 이유는 성능이 아니라 **학습 방식** 때문입니다.
> "다음에 올 단어"를 맞히도록 배운 적이 없으니까요.
>
> 👉 이 "미래를 못 보게 가리는 것"이 **causal mask** 이고,
> 과제에서 직접 구현하게 됩니다. Transformer의 핵심 장치 중 하나입니다.


Encoder의 대표 모델인 BERT 모델을 불러옵니다.

````python
from transformers import AutoTokenizer, AutoModelForMaskedLM

model_name = "bert-base-cased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForMaskedLM.from_pretrained(model_name).to(device)
````

BERT 모델을 이용하여 빈칸 맞추기(Masked Language Modeling)를 추론해봅니다. 예를 들어, I [MASK] to school. 이라는 문장에서 [MASK]에 들어갈 단어를 맞춘다고 하면 I go to school. 이 문장이 정답이 됩니다.

하지만, I went to school도 정답이 될 수 있습니다. 이처럼 [MASK]에 들어갈 단어는 여러가지가 될 수 있고, 모델의 학습에 따라 어떤 단어가 [MASK]에 들어갈지 결정됩니다.

이러한 특성을 이용하여 BERT 모델을 이용하여 빈칸 맞추기(`[MASK]`)를 추론해봅니다.

````python
# 우리가 맞출 문장 만들기
# tokenizer.mask_token = "[MASK]" 이 부분이 빈칸이 됨
sentence = f"I {tokenizer.mask_token} to school."

# Step 1에서 배운 인코딩 과정 수행 후
# BERT 모델에 문장(숫자형태)을 넣어서 예측 결과(logits) 얻기
encoded = tokenizer(sentence, return_tensors="pt", return_attention_mask=True).to(device)
outputs = model(**encoded)
logits = outputs.logits.squeeze(0)  # (seq_len, vocab_size)

# 이제 전체 logits 중에 [MASK] 토큰 자리에 어떤 토큰이 예측되었는지 확인하기.
input_ids = encoded.input_ids
mask_token_id = tokenizer.mask_token_id
mask_positions = (input_ids == mask_token_id).nonzero(as_tuple=False)

top_k = 5  # 상위 5개 후보 단어를 보고 싶다
all_token_candidates = []
for _, pos in mask_positions:
    pos = pos.item()  # 위치 숫자 꺼내기
    logits_at_pos = logits[pos]  # 해당 위치의 예측 점수
    probs = torch.softmax(logits_at_pos, dim=-1)  # 점수를 확률로 변환
    topk = torch.topk(probs, k=top_k)  # 확률이 높은 상위 5개 선택

    ids = topk.indices.tolist()   # 단어 ID
    scores = topk.values.tolist() # 확률 값

    # 단어 ID를 실제 단어(토큰)로 변환
    tokens = [tokenizer.convert_ids_to_tokens(tid) for tid in ids]

    # (단어, 확률) 형태로 묶어서 저장
    candidates = list(zip(tokens, scores))
    all_token_candidates.append(candidates)

# [MASK]에 들어갈 단어로 완성된 문장들을 저장할 리스트
restored_sentences = []
token_candidates = all_token_candidates[0]

# 후보 단어들을 하나씩 넣어서 문장을 만들어 보기
for tok, _ in token_candidates:
    new_ids = input_ids.clone()  # 원래 문장의 숫자 복사
    tok_id = tokenizer.convert_tokens_to_ids(tok)  # 후보 단어를 숫자로 변환
    new_ids[0, mask_positions[0, 1]] = tok_id      # [MASK] 위치에 후보 단어 ID 넣기
    text = tokenizer.decode(new_ids[0], skip_special_tokens=True)  # 다시 글자로 변환
    restored_sentences.append(text.strip())  # 앞뒤 공백 제거 후 저장

print("원본 문장:", sentence)
print("BERT가 예측한 문장들:")
for idx, sent in enumerate(restored_sentences, start=1):
    print("{}순위: {}".format(idx, sent))
````

> 🔍 **출력된 상위 5개 후보를 보면서 생각해볼 것**
>
> `I [MASK] to school.` 에서 BERT가 뽑은 후보들이 전부 문법적으로 자연스럽지 않나요?
> (`went`, `go`, `came`, `walked`, `returned` 같은 동사들)
>
> 주목할 점: 모델은 **품사를 따로 배운 적이 없습니다.** 문법 규칙을 입력받은 적도 없고요.
> 그저 대량의 문장에서 빈칸 맞추기를 반복했을 뿐인데, 그 자리에 동사가 와야 한다는 걸 알게 된 것입니다.
>
> 💡 **직접 실험해보세요**
> ```python
> sentence = f"The capital of France is {tokenizer.mask_token}."
> sentence = f"SSAFY is a {tokenizer.mask_token} program."
> sentence = f"이 영화는 정말 {tokenizer.mask_token}."   # bert-base-cased는 영어 모델이라 잘 못합니다
> ```
> 마지막 예시로 **모델이 학습한 언어를 벗어나면 어떻게 되는지** 꼭 확인해보세요.
> 한국어를 하려면 `klue/bert-base` 같은 한국어 모델이 필요합니다.


Only Decoder 모델의 대표인 GPT 모델을 이용하여 추론을 해보겠습니다.

GPT-2 모델을 불러옵니다.

````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name).to(device)
````

GPT-2 모델은 입력으로 토큰화된 텍스트를 받고, 그 뒤에 올 단어들을 예측(Next token Prediction)하는 것이 목표입니다.

아래 코드를 이용하여 스토리(입력 텍스트)의 뒷 내용을 생성해보겠습니다. (주의: 추론과정은 한 스텝별로 진행되다보니 시간이 오래 걸리는 편입니다)

````python
prompt = "Once upon a time in a small village, a curious child found a mysterious key."
inputs = tokenizer(prompt, return_tensors="pt").to(device)

with torch.no_grad():
    generated_ids = model.generate(**inputs, max_new_tokens=64)

output_tokens = tokenizer.decode(generated_ids.squeeze(), skip_special_tokens=True)
print(output_tokens)
````

> 💡 **생성 결과가 마음에 안 든다면 — `generate()` 의 손잡이들**
>
> 지금은 기본값(greedy)이라 문장이 반복되기 쉽습니다. 아래 옵션을 바꿔가며 실험해보세요.
>
> ```python
> generated_ids = model.generate(
>     **inputs,
>     max_new_tokens=64,
>     do_sample=True,        # 1등만 고르지 말고 확률에 따라 뽑기
>     temperature=0.8,       # 낮으면 보수적, 높으면 창의적 (0.7~1.0 권장)
>     top_p=0.9,             # 확률 상위 90% 안에서만 후보 선택
>     repetition_penalty=1.2,  # 같은 말 반복에 벌점
> )
> ```
>
> | 옵션 | 올리면 | 내리면 |
> |---|---|---|
> | `temperature` | 창의적·산만함 | 안정적·반복적 |
> | `top_p` | 다양한 단어 | 무난한 단어만 |
> | `repetition_penalty` | 반복 줄어듦 | 반복 늘어남 |
>
> 👉 여러분이 LLM API를 쓸 때 조절하는 그 파라미터들이 정확히 이것입니다.
> **모델을 바꾸지 않아도 출력 품질이 크게 달라집니다.** 프롬프트 엔지니어링만큼 중요한 영역입니다.
>
> ⚠️ `do_sample=True` 를 쓰면 실행할 때마다 결과가 달라집니다. 고정하려면 `torch.manual_seed()` 를 다시 호출하세요.

## 📌 이 실습의 핵심 정리

### 1. 모든 NLP는 3단 변환에서 시작합니다

```
"나는 학교에 간다"        문자열 (사람의 언어)
        ↓ 토크나이저
[2, 1523, 892, 431, 3]   정수 ID (기계가 세는 단위)
        ↓ nn.Embedding
[[0.1, -0.3, ...], ...]  실수 벡터 (의미를 담는 공간)
```

토크나이저는 **가중치가 아니라 단어장을 학습**하고, 임베딩은 **계산이 아니라 표 조회**입니다.

### 2. 시퀀스 모델의 발전은 '병목 제거'의 역사입니다

| 단계 | 해결한 문제 | 남은 병목 |
|---|---|---|
| **RNN** | 순서 정보 처리 | 긴 문장에서 앞부분이 사라짐 |
| **LSTM** | 게이트로 장기 기억 확보 | 여전히 **벡터 하나**에 문장 전체를 압축 |
| **Attention** | 인코더의 **모든** 상태를 참조 | 순차 처리라 **병렬화 불가** |
| **Transformer** | RNN을 버리고 Attention만 사용 | 시퀀스 길이의 제곱에 비례하는 연산량 |

### 3. shape을 따라가면 구조가 보입니다

```
input_ids       (L,)              토큰 L개
input_embeds    (L, 768)          각 토큰을 768차원 벡터로
hidden_states   (L, 1024)         RNN이 매 시점 만든 은닉 상태
h_n             (1, 1024)         마지막 상태 = 문맥 벡터
logits          (max_len, 30000)  각 스텝마다 3만 개 단어의 점수
```

> 에러가 나면 무조건 `print(tensor.shape)` 부터. 딥러닝 디버깅의 8할입니다.

### 4. 구조가 전부가 아닙니다

Step 2에서 만든 Seq2Seq와 Step 4의 Helsinki 번역 모델은 **구조가 같습니다.**
결과가 하늘과 땅 차이인 이유는 오직 하나, **학습 여부**입니다.

| 증상 | 원인 | 해결 |
|---|---|---|
| `IndexError: index out of range` (Embedding) | 토큰 ID가 `num_embeddings` 이상 | `len(tokenizer.get_vocab())` 로 크기 맞추기 |

| 같은 단어만 반복 생성 | greedy decoding + 미학습 | `do_sample=True`, `repetition_penalty` |

## ➡️ 과제에서 이어서 배울 것

이번 실습의 Attention은 **디코더가 인코더를 참조하는(cross)** 형태였습니다.
과제에서는 **자기 자신을 참조하는 Self-Attention** 과, 이를 여러 관점으로 나눈
**Multi-Head Attention** 을 직접 구현해 GPT-2를 완성합니다.

> 오늘 손으로 짠 `attention_weights = softmax(score)`, `context = weights @ values` 이 두 줄이
> 그대로 Transformer의 심장이 됩니다.
