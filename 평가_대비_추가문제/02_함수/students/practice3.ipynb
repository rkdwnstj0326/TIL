## 회문 판별

> 회문 또는 팰린드롬은 거꾸로 읽어도 제대로 읽는 것과 같은 문장이나 낱말, 숫자, 문자열 등을 말한다. 
> 
> 입력으로 짧은 영어단어 word가 주어질 때, 해당 단어가 회문이면 True 회문이 아니면 False를 반환하는 함수를 작성하시오.
> 
> 이때, 반복문(`while`)과 재귀 함수를 사용해서 각각 작성하시오.

---

예시)

```python
is_pal_while('tomato') #=> False
is_pal_while('racecar') #=> True
is_pal_recursive('azza') #=> True
```

````python
# while

def is_pal_while(word):
    while len(word) > 1: # 현재 단어의 길이가 1보다 큰 동안 계속 반복해라
        # 양쪽을 비교하려면 최소한 글짜가 2개 이상은 있어야 댐 so. 길이가 1이 되면 반복을 끝냄
        if word[0] == word[-1]:
            word = word[1:-1] #문자열 슬라이싱 + 오른쪽의 word[1:-1]로 새로운 문자열을 만들어서 그 결과를 다시 word에 저장
        else: #if word[0] == word[-1]: 에서 양쪽 글자가 같지 않으면 else로 이동
            return False
    return True
````

````python
# 해당 코드를 통해 올바른 결과가 나오는지 확인하시오.
print(is_pal_while('tomato'))
print(is_pal_while('racecar'))
print(is_pal_while('azza'))
````

````python
# recursive
# 재귀함수는 자기자신을 다시 호출하기 때문에 반드시 멈추는 조건이 있어야 함!(종료조건)

def is_pal_recursive(word): 
    # 1. 종료조건 선언
    if len(word) <= 1:
        return True #현재 단어의 길이가 1이하라면 더 이상 비교할 필요가 없으니 true를 반환하고 끝내자
    
    # 2. 양 끝이 같으면 => 다음 subword를 넣어 함수 호출
    if word[0] == word[-1]: 
        return is_pal_recursive(word[1:-1])
    # 2-1. 다르면 => False
    else:
        return False # 첫번째 글자와 마지막 글자가 다르면 회문이 아니어서 false
    
````

````python
# 해당 코드를 통해 올바른 결과가 나오는지 확인하시오.
print(is_pal_recursive('tomato'))
print(is_pal_recursive('racecar'))
print(is_pal_recursive('azza'))
````
