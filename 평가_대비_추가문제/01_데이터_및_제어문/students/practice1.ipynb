## 갯수 구하기

> 주어진 리스트의 요소는 학생 이름으로 구성되어 있다. 학생들의 수를 출력하시오.

---

```
[출력 예시]
3
```

````python
students = ['김철수', '이영희', '조민지']

# 아래에 코드를 작성하시오.

print(len(students))
#len : 리스트 안에 있는 요소의 개수를 세는 함수임
#여기서 int를 쓰면 안됨!!! int는 리스트 전체를 정수로 바꾸려고 하기 때문에 오류남
````

## 득표수 구하기

> 주어진 리스트는 반장 선거 투표 결과이다. 이영희의 총 득표수를 출력하시오.

---

```
[출력 예시]
4
```

````python
students = ['이영희', '김철수', '이영희', '조민지', '김철수', '조민지', '이영희', '이영희']

# 아래에 코드를 작성하시오.

# print(len('이영희')) -> 이렇게 하면 득표수를 센게 아니라 문자열의 글자수를 센거임!! 이 영 희 이렇게 3글자여서 3이 출력됨

print(students.count('이영희'))
# students.count를 해야지 리스트에 몇번 들어있는지 셀 수 있음
````

## 최댓값 구하기

> 주어진 리스트의 요소 중에서 최댓값을 출력하시오.

---
```
[출력 예시]
22
```

````python
numbers = [7, 10, 22, 4, 3, 17]

# 아래에 코드를 작성하시오.

print(max(numbers))
# 최댓값: max()
# 최솟값: min()
# 중간값: median()
````

## 최댓값과 등장 횟수 구하기

> 주어진 리스트의 요소 중에서 최댓값과 등장 횟수를 출력하시오.

---

```
[출력 예시]
22 3
```

````python
numbers = [7, 10, 22, 7, 22, 22]

# 아래에 코드를 작성하시오.

print(max(numbers), numbers.count(max(numbers)))
# max(numbers): 가장 큰 값인 22 찾음
# numbers.count(max(numbers)): 먼저 max로 22 찾고 -> numbers.count(22)가 되어 리스트 안에 22가 몇번 있는지 찾음

# 방법2 : 코드를 조금 더 읽기 쉽게 쓰려면 변수에 저장하기
# max_number = max(number)로 정의하면 더 쉬움!
````

## 'a'가 싫어


> 입력으로 짧은 영단어 word가 주어질 때, 해당 단어에서 'a'를 모두 제거한 결과를 출력하시오.

---
```
[입력 예시]
apple

[출력 예시]
pple
```

````python
word = input()

# 아래에 코드를 작성하시오.

print(word.replace('a',''))
# 문자열.replace('바꿀 문자', '새로운 문자')
# a를 찾아서 ''로 바꾸라는 뜻
````

## 단어 뒤집기

> 입력으로 짧은 영어단어 word가 주어질 때, 해당 단어를 역순으로 뒤집은 결과를 출력하시오.

---
```
[입력 예시]
apple

[출력 예시]
elppa
```

````python
word = input()

# 아래에 코드를 작성하시오.

print(word[::-1])
# 슬라이싱 : 문자열[시작:끝:간격]
#[::-1] : 시작과 끝은 전체 범위로 두고(시작위치, 끝위치 생략한게 문자열 전체를 뜻함), 뒤에서 앞으로 한 글자씩 이동한다는 뜻
````
