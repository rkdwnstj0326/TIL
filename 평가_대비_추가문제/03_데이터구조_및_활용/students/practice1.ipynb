## 복잡한 리스트의 합

> 2차원 리스트를 반복하는 방법을 알아봅시다. 


* 주어진 아래의 리스트를 반복하여 숫자의 합을 반환하시오.

---
**예시 입력**

```python
numbers = [
    [1, 4],
    [10, 5],
    [20, 30]    
]

```

**예시 출력**
```python
sum_list([[1, 4], [10, 5], [20, 30]]) # 70
```

### for 문을 활용하여 풀이하기

```python
for element in list:
```

````python
# 아래에 코드를 작성하세요.
def sum_list(numbers):
    result = 0 #숫자를 하나씩 더해서 저장할 변수! 처음에는 아무 숫자도 더하지 않았으니까 0부터 시작함

    for number in numbers: # 작은 리스트를 하나씩 꺼내는 코드!
        for n in number: #number안의 숫자를 또 하나씩 꺼내는거임
            result = result + n #지금까지 더한 값 result에 새로 꺼낸 숫자 n을 더하고 그 결과를 다시 result에 저장한다.

    return result
````

````python
print(sum_list([[1, 4], [10, 5], [20, 30]]))
````

### Index로 접근하여 풀이하기

```python
for element in range(len(list)):
```

````python
# 아래에 코드를 작성하세요.
def sum_list_index(numbers):
    result = 0

    for i in range(len(numbers)): #len(numbers) : 3 -> [1,4][10,5][20,30]
        # range(3) -> 0,1,2 / so. 첫번째 i=0, 두번째 i=1, 세번째 i=2
        # numbers의 인덱스인 0,1,2를 i에 하나씩 넣으면서 반복하겠다.
        for j in range(len(numbers[i])):
            #현재 선택한 내부 리스트의 인덱스를 j에 하나씩 넣으면서 반복하겠다.
            result = result + numbers[i][j] #numbers[i][j] : 실제 숫자 하나를 가져온다
            # 가져온 숫자를 합계에 누적한다.

    return result
````

````python
print(sum_list_index([[1, 4], [10, 5], [20, 30]]))
````

### while 문을 활용하여 풀이하기

```python
while ____:
```

````python
## 아래에 코드를 작성하세요.
def sum_list_while(numbers):
    result = 0 #전체 합계를 저장할 변수
    i=0 #바깥쪽 리스트의 인덱스

    while i < len(numbers): #len(numbers):3 -> i는 0,1,2일 때만 반복
        j=0 #새로운 내부 리스트를 확인할 때 마다 인덱스를 0으로 초기화
        while j < len(numbers[i]): #현재 내부 리스트의 인덱스가 범위 안에 있는 동안 반복
            result = result + numbers[i][j]
            j = j+1 #내부 리스트의 다음 숫자로 이동

        i=i+1

    return result
````

````python
i = 0               # 1. 시작값

while i < 3:         # 2. 반복 조건
    print(i)
    i = i + 1        # 3. 값 변화
````

````python
print(sum_list_while([[1, 4], [10, 5], [20, 30]]))
````

## 시험 점수

> 2차원 배열


* A반 학생들의 점수는 아래와 같고, students 리스트에 저장되어 있다.

    * A학생(국어 100점, 수학 80점, 영어 100점)
    * B학생(국어 90점, 수학 90점, 영어 60점)

### 학생별 출력

**예시 입력**

students = [

````python
students = [
    [100, 80, 100],
    [90, 90, 60],
    [80, 80, 80],
]
````

````python
# 아래에 풀이를 작성하세요. 함수 정의는 필요없습니다.
# result = 0 -> 얘가 여기있으면 학생별로 초기화가 안됨! 모든 학생 점수가 계속 누적됨
for student in students:
    result = 0
    for score in student:
        result = result + score
        
    print(result)
````

### 과목별 출력

* 아래의 리스트를 반복하여 **과목별 총합**을 순서대로 `출력`하시오.
* `sum` 함수 사용 금지

---
**예시 입력**

```python
students = [
 [100, 80, 100],
 [90, 90, 60],
 [80, 80, 80]
]
```

**예시 출력**
```
270
250
240
```

````python
students = [
    [100, 80, 100],
    [90, 90, 60],
    [80, 80, 80],
]
````

````python
# 아래에 풀이를 작성하세요. 함수 정의는 필요없습니다.
# [0][0] [1][0] [2][0] 이런식으로 더해야함
for i in range(len(students[0])): #range를 해야하는 이유 : 숫자를 차례로 만듬 0,1,2로
    # students[0] : 첫번째 학생의 국어 수학 영어 점수
    # len(students[0]) : 점수가 3개 들어있어서 3!
    # range(3) : 숫자 0,1,2로 만듬! 즉 인덱스가 필요하기 때문!!!
    # -> for i in range(3) : 반복할 때 i에는 0->국어 / 1->수학 / 2->영어 이렇게 들어감
    # 과목을 국어 수학 영어 순서로 하나씩 선택한다는 뜻

    result = 0 #현재 과목의 점수를 더해 저장할 변수

    for j in range(len(students)):
        # len(students) : 학생 수가 3명이어서 3!
        # range(3) : 인덱스 번호를 만듬
        # -> 학생 인덱스 0,1,2를 j에 하나씩 넣으면서 모든 학생을 반복한다.
        # 학생을 a학생, b학생, c학생 순서로 하나씩 선택한다는 뜻

        result = result + students[j][i]
        # 현재까지 더한 값에 현재 학생의 현재 과목 점수를 추가하는 것

    print(result)

# i : 과목번호
# j : 학생번호
````
