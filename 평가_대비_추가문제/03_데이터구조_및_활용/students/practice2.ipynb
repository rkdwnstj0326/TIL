## 모든 위치

> 주어진 문자열(text)에서 제시된 알파벳(alphabet)의 등장 위치를 리스트로 반환하시오.
> 해당 알파벳이 등장하지 않으면, -1을 반환하시오.

---

예시)

```python
print(my_find('apple', 'p')) # [1, 2]
print(my_find('a', 'p'))     # -1
```


````python
# 문자열 전체를 돌면서 찾는 글자의 위치를 모두 모은 뒤, 찾았으면 위치 리스트를반환하고 못찾으면 -1 반환하는 함수

def my_find(text, alphabet): #my_find라는 함수를 만들고, 두가지 값을 받겠다.
    result = [] # 찾은 글자의 위치 번호를 저장할 빈 리스트
    # apple에서 p는 1번과 2번 위치에 있어서 result = [1,2]가 되어야 함.

    for idx, char in enumerate(text): #enumerate()는 반복할 때 인덱스 번호와 실제 값을 동시에 꺼내주는 함수
        if alphabet in char: #현재 문자 char안에 찾는 글자 alphabet이 들어있는지 확인
            result.append(idx) #append() : 리스트 맨 뒤에 값을 추가하는 기능

    if result: #찾는 위치가 하나라도 있다면
        return result #result 반환
    else:
        return -1
````

````python
# 해당 코드를 통해 올바른 결과가 나오는지 확인하시오.
print(my_find('apple', 'p'))
print(my_find('a', 'p'))
````

## 출석 체크

> 주어진 학생 n과 출석한 학생명부 students 문자열이 있다. 결석한 학생들로 구성된 문자열을 반환하시오.
>
> n이 7일 때, 1 2 3 4 5 6 7의 출석 번호가 부여되고,
>
> '1 3 5'는 출석한 학생 명부이다. 
>
> 즉, 결석한 학생 명부 '2 4 6 7'을 return 해야 한다.

---

예시)

```python
print(check(7, '1 3 5')) # 2 4 6 7
```


````python
# 전체 학생 번호 1부터 n까지 확인하면서, 출석 명부에 없는 번호만 모으면 됨

def check(n, students): #check라는 이름의 함수를 만드는 코드
    students = list(map(int, students.split()))
    result = []

    for i in range(1, n + 1):
        if i not in students:
            result.append(str(i))

    return ' '.join(result)
````

````python
# 해당 코드를 통해 올바른 결과가 나오는지 확인하시오.
print(check(7, '1 3 5')) # 2 4 6 7
````
