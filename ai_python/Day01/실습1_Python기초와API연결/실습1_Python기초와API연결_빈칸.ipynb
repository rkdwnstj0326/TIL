## Step 1: 변수와 자료형

### Concept Check: 변수는 데이터를 담는 공간

변수는 데이터를 저장해 두는 **칸**입니다. `=` 는 오른쪽 값을 먼저 계산해 왼쪽 변수에 담는 대입 연산입니다.

```python
score = 90        # score 칸에 90을 담음
score = score + 10  # 오른쪽(90+10)을 먼저 계산해 다시 score에 담음 , 100
```

Python 은 `score = 90` 처럼 타입을 미리 적지 않고 값을 담습니다. 담는 값에 따라 자료형이 정해지는 **동적 타이핑**이기 때문입니다.

**기본 자료형 네 가지**

| 자료형 | 뜻 | 예시 |
| --- | --- | --- |
| `int` | 정수 | `10`, `-3`, `0` |
| `float` | 실수(소수점) | `3.14`, `-0.5` |
| `str` | 문자열(글자) | `"안녕"`, `"AI"` |
| `bool` | 참 / 거짓 | `True`, `False` |

같은 `5` 라도 `5`(int)와 `"5"`(str)는 다릅니다. `type(변수)` 로 자료형을 확인합니다.

**변수 이름 규칙**: 영문 또는 `_` 로 시작하고 숫자로는 시작할 수 없습니다. 공백 없이 `_` 로 단어를 잇고(`user_name`), `if`,`for`,`True` 같은 예약어는 쓸 수 없습니다.

**Java 와의 차이**

| 구분 | Java | Python |
| --- | --- | --- |
| 변수 선언 | `int n = 10;` (타입 명시) | `n = 10` (타입 자동) |
| 타입 시스템 | 정적 타입 (컴파일 시 결정) | 동적 타입 (실행 시 결정) |
| 리스트/배열 | `int[] a = {1, 2, 3};` | `a = [1, 2, 3]` |
| 맵/딕셔너리 | `Map<String,Integer> m = new HashMap<>();` | `m = {"a": 1}` |
| 문장 종결 | 세미콜론 `;` 필수 | 줄바꿈 (세미콜론 불필요) |
| 코드 블록 | 중괄호 `{ }` | 들여쓰기(indent) |

````python
# 변수에 값을 담고, type() 으로 자료형을 확인해 봅니다
age = 27
height = 175.5
name = "김싸피"
is_student = True

print(age, type(age))            # 27 <class 'int'>
print(height, type(height))      # 175.5 <class 'float'>
print(name, type(name))          # 김싸피 <class 'str'>
print(is_student, type(is_student))  # True <class 'bool'>

# 사칙연산: 나누기(/)는 항상 float, 몫(//)과 나머지(%)에 주의
print(age / 2)    # 13.5
print(age // 2)   # 13  (몫)
print(age % 2)    # 1   (나머지)
print(age ** 2)   # 729 (제곱)
````

### Concept Check: 형변환과 자주 만나는 에러

엑셀,CSV 에서 읽은 값은 숫자처럼 보여도 대부분 **문자열(`str`)** 입니다. 계산하려면 먼저 형변환을 해야 합니다.

- `int("27")`은 `27`, `float("3.14")`는 `3.14`, `str(27)`은 `"27"`로 변환
- `int(3.99)`는 `3` (반올림 없이 **소수점 아래를 버림**)

형변환이 실패하면 에러가 납니다. 빨간 메시지를 읽고 무엇이 문제인지 파악하는 습관이 중요합니다.

- **`ValueError`**: `int("3.14")` - 소수점이 있는 문자열은 정수로 변환할 수 없음, `float`로 변환한 뒤 `int` 사용
- **`TypeError`**: `"김싸피" + 27` - `str`과 `int`는 더할 수 없음, `"김싸피" + str(27)` 또는 f-string 사용

> **f-string** 은 문자열 안에 변수를 넣는 방법입니다: `f"{name}님은 {age}살"`.

````python
# 형변환이 실패하는 두 가지 상황을 눈으로 확인합니다 (에러가 정상입니다)
try:
    int("3.14")
except ValueError as e:
    print("ValueError:", e)

try:
    result = "김싸피" + 27
except TypeError as e:
    print("TypeError:", e)

# 올바른 해결
print(int(float("3.14")))   # 3
print("김싸피" + str(27))     # 김싸피27
````

## Step 2: 자료구조 - `list`

### Concept Check: 순서로 찾는 자료구조, list

**`list`(리스트)** 는 여러 값을 **순서대로** 묶어 담는 자료구조입니다. 대괄호 `[ ]` 로 만들고, 각 칸에는 `0` 부터 시작하는 **인덱스**가 매겨집니다.

```python
scores = [90, 85, 70, 60, 50]
#          0   1   2   3   4    , 앞에서부터
#         -5  -4  -3  -2  -1    , 뒤에서부터
```

- **인덱싱** `scores[0]`: 인덱스로 특정 값에 접근. 첫 값은 `[0]`, 마지막 값은 `[-1]`.
- **슬라이싱** `scores[1:4]`: 인덱스 1, 2, 3의 값을 선택. **끝 인덱스는 포함하지 않습니다.**
- `len(scores)` 길이, `sum(scores)` 합계.

`list` 는 한 곳에 서로 다른 자료형을 섞어 담을 수 있습니다(`[1, "AI", 3.14]`).

````python
scores = [90, 85, 70, 60, 50]

print(scores[0])     # 90  (첫 번째)
print(scores[-1])    # 50  (마지막)
print(scores[1:4])   # [85, 70, 60]  (인덱스 1,2,3 - 4는 제외)
print(scores[:3])    # [90, 85, 70]  (처음부터 인덱스 2까지)
print(scores[::2])   # [90, 70, 50]  (한 칸씩 건너뛰기)

# 슬라이싱은 범위를 넘어도 에러 없이 가능한 만큼만 반환
print(scores[0:99])  # [90, 85, 70, 60, 50]
# 인덱싱은 없는 칸에 접근하면 IndexError
# print(scores[99])  # , 주석을 풀면 IndexError
````

### Concept Check: 이름으로 찾는 자료구조, dict

**`dict`(딕셔너리)** 는 **키(key)**와 **값(value)**을 쌍으로 담습니다. 중괄호 `{ }` 안에 `키: 값` 형태로 쓰고, **키로 값을 찾습니다.**

```python
person = {"name": "김싸피", "age": 27}
person["name"]   # "김싸피"
```

키는 중복될 수 없습니다. `list` 와 비교하면 이렇게 나뉩니다.

| 구분 | `list` | `dict` |
| --- | --- | --- |
| 모양 | `[v, v, v]` | `{k: v, k: v}` |
| 찾는 법 | 순서(인덱스) | 이름(키) |
| 쓰임 | 점수 목록처럼 순서 있는 데이터 | 한 사람의 정보처럼 이름표 붙은 데이터 |

> 참고: `dict` 는 입력한 **순서를 유지합니다**(Python 3.7+). 그래서 `list(d.keys())` 는 넣은 순서대로 키를 돌려줍니다.

````python
person = {"name": "김싸피", "age": 27, "job": "학생"}

# 접근
print(person["name"])              # 김싸피
print(person.get("hobby", "없음"))  # 없음  (없는 키 , 기본값)
print("age" in person)             # True  (키 존재 여부)

# 추가,수정,삭제
person["hobby"] = "독서"    # 새 키 추가
person["age"] = 28         # 기존 값 수정
print(person.pop("job"))   # 학생  (제거하며 값 반환)

# 순회
for key, value in person.items():
    print(key, "=", value)
````

## Step 4: 조건문

### Concept Check: if / elif / else

조건문은 참/거짓에 따라 다른 코드를 실행합니다. 위에서부터 검사해 **처음 참이 되는 조건의 코드만** 실행하고 빠져나옵니다.

```python
if score >= 90:
    print("A")
elif score >= 80:   # 90 미만이면서 80 이상?
    print("B")
else:
    print("그 외")
```

어디까지가 `if` 에 속하는지는 **들여쓰기(보통 4칸)** 로 정해집니다. Java 의 중괄호 `{ }` 자리를 들여쓰기가 대신합니다.

**비교,논리 연산자**

| 연산자 | 의미 | | 연산자 | 의미 |
| --- | --- | --- | --- | --- |
| `==` | 같음 (값+타입) | | `and` | 둘 다 참 |
| `!=` | 다름 | | `or` | 하나라도 참 |
| `>=` `<=` | 이상 / 이하 | | `not` | 반대 |

`==` 는 값과 타입을 함께 비교합니다. `5 == "5"` 는 `False` 입니다.

**거짓으로 평가되는 값**: 숫자 `0`, 빈 문자열 `""`, 빈 리스트 `[]`, `None`. 그래서 `if not scores:` 로 "리스트가 비었는지" 를 검사할 수 있습니다.

````python
# 비교,논리 연산자, 그리고 빈 값의 참/거짓
age = 25
is_student = False

print(age >= 20 and not is_student)   # True
print(age < 10 or age >= 65)          # False
print(20 <= age < 30)                 # True  (Python 만의 연쇄 비교)
print(5 == "5")                       # False (숫자와 문자열은 다른 값)

scores = []
if not scores:        # 빈 리스트는 거짓 , not 으로 뒤집으면 참
    print("점수가 비어 있습니다")

name = "김싸피"
if name:              # 비어있지 않은 문자열은 참
    print("이름 있음")
````

## Step 5: 반복문

### Concept Check: for 와 range

`for` 는 `list`, 문자열 같은 묶음에서 값을 차례대로 꺼내 반복합니다.

```python
for s in [90, 85, 70]:
    print(s)        # 90, 85, 70 차례로
```

정해진 횟수만큼 돌릴 때는 `range` 를 씁니다.

| 표현 | 결과 |
| --- | --- |
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(1, 5)` | 1, 2, 3, 4 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -1)` | 10, 9, ..., 1 (역순) |

- `break` : 루프를 즉시 빠져나감
- `continue` : 이번 회차만 건너뛰고 다음으로
- `for`-`else` : `break` 없이 반복이 모두 끝났을 때만 `else` 가 실행됨

**반복하며 같은 리스트에서 원소를 삭제하면** 인덱스가 밀려 일부를 건너뜁니다. 조건에 맞는 값만 남길 때는 **새 리스트에 담는** 방식을 사용합니다.

````python
# break 와 continue
for i in range(10):
    if i == 3:
        break        # 3에서 멈춤
    print("break 예시:", i)   # 0, 1, 2

for i in range(5):
    if i == 2:
        continue     # 2만 건너뜀
    print("continue 예시:", i)  # 0, 1, 3, 4

# for-else: break 없이 끝까지 돌면 else 가 실행됨
for i in range(3):
    print("for-else 예시:", i)
else:
    print("루프 정상 완료")     # break 로 빠져나오지 않았으므로 실행
````

````python
scores = [90, 85, 70, 60]

# 1. 반복문으로 scores 의 합계를 구해 total 에 담으세요.
total = 0

for score in scores:
    total += score

assert total == 305, "합계가 올바르지 않습니다."

# 2. scores 의 평균을 구해 average 에 담으세요.
average = total / len(scores)

assert round(average, 2) == 76.25, "평균이 올바르지 않습니다."

# 3. 1~10 중 짝수만 더한 값을 even_sum 에 담으세요.
even_sum = 0

for number in range(1, 11):
    if number % 2 == 0:
        even_sum += number

assert even_sum == 30, "짝수 합이 올바르지 않습니다."

# 4. scores 에서 80점 이상인 값만 모은 리스트를 passed 에 담으세요.
passed = []

for score in scores:
    if score >= 80:
        passed.append(score)

assert passed == [90, 85], "필터 결과가 올바르지 않습니다."

print("TODO 8 통과!")
````

## Step 6: 함수

### Concept Check: def 와 return

함수는 반복되는 작업에 **이름을 붙여 묶어 둔 것**입니다. 한 번 만들면 이름으로 계속 호출해 재사용합니다.

```python
def average(numbers):
    return sum(numbers) / len(numbers)

average([90, 85, 70])   # 81.66...
```

- **기본값 매개변수** `def greet(name, greeting="안녕하세요")` - 호출 때 인자를 생략하면 기본값이 쓰입니다.
- **여러 반환값** `return min_v, max_v` - 튜플로 반환되며 `a, b = ...` 로 풀 수 있습니다.
- **가변 인자** `def add(*nums)` - 인자 개수가 정해지지 않을 때.
- `return` 이 없는 함수는 `None` 을 반환합니다.

**스코프**: 함수 안에서 만든 **지역변수**는 함수가 끝나면 사라져 밖에서 접근할 수 없습니다(`NameError`). 함수 밖의 **전역변수**는 함수 안에서 읽을 수 있습니다.

| 변수 종류 | 생성 위치 | 접근 범위 | 생명 주기 |
| --- | --- | --- | --- |
| 전역변수 | 함수 외부 | 모든 곳 | 프로그램 시작~종료 |
| 지역변수 | 함수 내부 | 함수 내부만 | 함수 호출~반환 |

단, `list`,`dict` 는 참조형이라 함수에 넘겨 안에서 바꾸면 **원본도 변경됩니다.**

**Java 와의 차이**

| 구분 | Java | Python |
| --- | --- | --- |
| 조건문 | `if (x > 0) { ... }` | `if x > 0:` (콜론+들여쓰기) |
| else if | `else if (...) { }` | `elif ...:` |
| for 반복 | `for (int i=0; i<n; i++)` | `for i in range(n):` |
| 함수 정의 | `int add(int a, int b){ return a+b; }` | `def add(a, b): return a + b` |
| 반환 타입 | 명시 필요 (`int`, `void`) | 불필요 (자동 추론) |
| 불리언 | `boolean`, `true` / `false` | `bool`, `True` / `False` |

````python
# 여러 반환값, 가변 인자, 참조형 동작을 함께
def get_min_max(numbers):
    return min(numbers), max(numbers)

min_v, max_v = get_min_max([90, 85, 70])   # 언패킹
print(min_v, max_v)                        # 70 90

def add(*numbers):       # 가변 인자
    return sum(numbers)
print(add(1, 2, 3))      # 6

def add_score(lst):      # 참조형: 원본을 직접 수정
    lst.append(60)
target = [90, 85]
add_score(target)
print(target)            # [90, 85, 60]  , 원본도 변경됨

# 스코프: 지역변수는 함수 밖에서 접근 불가
global_var = 100

def my_func():
    local_var = 50       # 지역변수
    print(global_var)    # 100  (전역변수는 읽기 가능)
    print(local_var)     # 50

my_func()
try:
    print(local_var)     # 함수 밖에서는 접근 불가
except NameError as e:
    print("NameError:", e)
````

### TODO 10: 문자열 속 숫자의 합

**요구사항**
- 문자열에 섞인 숫자만 골라 더하는 함수 `sum_of_digits` 를 완성합니다.
- 예: `sum_of_digits("a1b2c3")` , `6` (1 + 2 + 3)

**제한조건**
- 정답 확인용 `assert` 줄은 수정하지 않습니다.

**힌트**
- 글자 하나씩 도는 것은 `for ch in s`.
- 숫자인지 확인은 `ch.isdigit()`, 글자를 정수로 바꾸는 것은 `int(ch)`.

````python
# 문자열에 섞인 숫자만 골라 더해 돌려주는 함수
def sum_of_digits(s: str) -> int:
    total = 0

    for char in s:
        if char.isdigit():
            total += int(char)

    return total


assert sum_of_digits("a1b2c3") == 6, "1+2+3 = 6 이어야 합니다."
assert sum_of_digits("hello") == 0, "숫자가 없으면 0 이어야 합니다."
assert sum_of_digits("2024년") == 8, "2+0+2+4 = 8 이어야 합니다."

print("TODO 10 통과!")
````

## Step 7: API 와 HTTP - `requests`

### Concept Check: API 요청과 응답

API 는 프로그램끼리 요청과 응답을 주고받는 약속입니다.

- **API 문서**: 보낼 수 있는 요청의 종류와 형식을 정의
- **요청(Request)**: 클라이언트가 서버로 보내는 메시지
- **응답(Response)**: 서버가 돌려주는 결과

서버 내부 구현을 몰라도 요청과 응답 형식만 알면 사용할 수 있습니다. 우리 코드(클라이언트)는 **요청을 보내고 응답을 받습니다.**

**HTTP 메서드**: 데이터를 *가져올* 때는 `GET`(본문 없이 주소에 조건을 실음), 데이터를 *보낼* 때는 `POST`(본문에 JSON 을 실음)를 씁니다. 조회에는 `GET` 을, LLM 호출에는 `POST` 를 사용합니다.

**상태 코드(status code)** 는 요청 처리 결과를 나타내는 숫자입니다.

| 코드 | 의미 | 구분 |
| --- | --- | --- |
| `200` | 성공 | - |
| `404` | 자원 없음 | 클라이언트 요청 오류 |
| `500` | 서버 내부 오류 | 서버 오류 |

> 상태 코드는 **첫 자리 숫자**로 종류가 나뉩니다. `2xx`는 요청 성공, `4xx`는 클라이언트 요청 오류, `5xx`는 서버 오류를 의미합니다.

**`response` 객체의 주요 속성**

| 속성 | 타입 | 의미 |
| --- | --- | --- |
| `.status_code` | `int` | HTTP 상태 코드 (예: `200`) |
| `.text` | `str` | 응답 본문 (원본 **문자열**) |
| `.json()` | 메서드 | 본문을 파이썬 `dict`, `list`로 변환 |
| `.headers` | `dict` | 응답 헤더 |
| `.elapsed` | `timedelta` | 요청부터 응답까지 걸린 시간 |
| `.url` | `str` | 실제 요청된 URL |

`.text`는 문자열이므로 `["name"]`처럼 키로 접근할 수 없습니다. JSON 응답을 파이썬 객체로 다룰 때는 `.json()`을 사용합니다.

### Concept Check: JSON 과 파이썬 자료형의 대응

JSON은 프로그램끼리 데이터를 주고받을 때 사용하는 텍스트 기반 형식이며, 파이썬 자료형과 다음처럼 대응합니다.

| JSON | 파이썬 | 접근 방법 |
| --- | --- | --- |
| `{ ... }` | `dict` | `d["key"]` |
| `[ ... ]` | `list` | `l[0]` |
| `"text"` | `str` | |
| `10`, `true` | `int`, `bool` | |
| `null` | `None` | |

API 응답은 보통 `dict` 안에 `dict`, `list` 안에 `dict` 처럼 **중첩(nested)** 돼 있습니다. 중첩 접근 방법은 Step 3 과 같습니다. **바깥쪽부터 안쪽까지** 키와 인덱스에 순서대로 접근합니다.

**`d["key"]` vs `d.get("key")`**

- `d.get("key")` - 없으면 `None` 반환. **없을 수도 있는** 선택적 필드에 사용.

API 응답은 사용자마다 일부 필드가 빠질 수 있으니, 선택적 필드에는 `get`을 사용합니다. `list` 인덱스도 범위를 벗어나면 `IndexError` 가 나므로 길이를 먼저 확인합니다.

- 중첩 접근: `user["address"]["city"]`, `user["company"]["name"]`.
- 표 만들기: `pd.DataFrame(rows)`.

````python
import pandas as pd

# 각 사용자에서 name, email, city, company 를 뽑아 rows 에 담고, df 로 만드세요.
rows = []

for user in data:
    row = {
        "name": user["name"],
        "email": user["email"],
        "city": user["address"]["city"],
        "company": user["company"]["name"],
    }

    rows.append(row)

df = pd.DataFrame(rows)

assert len(rows) == len(data), "rows 개수가 사용자 수와 다릅니다."
assert {"name", "email", "city", "company"}.issubset(df.columns), "필요한 열이 없습니다."
assert len(df) == len(data) and df["city"].notna().all(), "DataFrame 이 올바르지 않습니다."

print("TODO 13 통과!")

print(df.info())
print(df["city"].value_counts())
df.head()
````

### Concept Check: LLM 호출도 요청 , JSON 응답

ChatGPT, Claude 같은 LLM 도 AI 서버에서 동작하므로, 호출 방식은 Step 7 의 요청-응답과 같습니다. 질문에 어떤 자료(컨텍스트)를 덧붙여 보내느냐가 달라집니다.

같은 질문이라도 함께 넣은 자료에 따라 응답 품질이 달라집니다.

| 시나리오 | 전달 내용 | 응답 경향 |
| --- | --- | --- |
| 컨텍스트 없음 | "서울 사용자 이메일은?" | 답변 근거 없음 |
| 관련 컨텍스트 포함 | 질문과 사용자 표 | 제공한 자료를 바탕으로 답변 |

모델 내부 지식만으로는 *우리 데이터*(사용자 목록)를 알 수 없습니다. 모델이 모르는 자료는 질문에 덧붙여 전달해야 합니다. 검색으로 관련 자료를 찾아 컨텍스트에 포함하는 방식이 **RAG**이고, 모델이 도구를 선택해 작업하도록 구성한 방식이 **Agent**입니다.

아래 `mock_llm` 은 실제 LLM API 응답과 **같은 JSON 모양**을 흉내 낸 모의 함수입니다. API 키 가 없어도 흐름을 익힐 수 있도록 준비했습니다.

````python
def mock_llm(prompt: str) -> dict:
    """실제 LLM API 응답과 같은 JSON 구조를 흉내 낸 모의 함수 (API 키 불필요)."""
    answer = "도시가 'Gwenborough' 인 사용자는 Leanne Graham 이며, 이메일은 Sincere@april.biz 입니다."
    return {
        "model": "mock-model",
        "usage": {"prompt_tokens": 50, "completion_tokens": 25, "total_tokens": 75},
        "choices": [
            {
                "index": 0,
                "message": {"role": "assistant", "content": answer},
                "finish_reason": "stop",
            }
        ],
    }
````

### TODO 14: 컨텍스트 문자열 만들고 모델 호출하기

**요구사항**
- 표의 각 행을 `- 이름 (도시), 이메일` 한 줄로 만들어 컨텍스트 문자열 하나로 합친 뒤, 질문과 묶어 `mock_llm` 을 호출합니다.

**제한조건**

**힌트**
- 한 줄 만들기: `f"- {r['name']} ({r['city']}), {r['email']}"`.
- 줄바꿈으로 잇기: `"\n".join(lines)`. 모델 호출: `mock_llm(prompt)`.

````python
# 가짜 모델 응답을 만들어 주는 함수
def mock_llm(prompt):
    return {
        "choices": [
            {
                "message": {
                    "content": "Leanne Graham, Sincere@april.biz"
                },
                "finish_reason": "stop"
            }
        ],
        "usage": {
            "total_tokens": 75
        }
    }


# 1. 각 행을 "- 이름 (도시), 이메일" 형태 문자열로 만들어
# 줄바꿈으로 이어 context 에 담으세요.
context = "\n".join(
    f"- {r['name']} ({r['city']}), {r['email']}"
    for r in rows
)

question = "도시가 'Gwenborough' 인 사용자의 이름과 이메일을 알려줘."

prompt = f"""다음 자료를 참고해 답해줘.
[자료]
{context}

[질문]
{question}"""

# 2. mock_llm 에 prompt 를 넣어 호출하고 결과를 result 에 담으세요.
result = mock_llm(prompt)

assert "\n" in context and context.count("\n") + 1 == len(rows), \
    "컨텍스트 줄 수가 사용자 수와 다릅니다."

assert isinstance(result, dict) and "choices" in result, \
    "모델 응답이 올바르지 않습니다."

print("TODO 14 통과!")
````

### Concept Check: 요청 본문과 응답 JSON 구조

실제 LLM API 는 `POST` 로 JSON 본문을 받습니다. 아래 코드는 요청 형태를 설명하기 위한 의사코드이며 실행하지 않습니다.

```python
body = {
    "model": "some-model",                                 # 사용할 모델 이름
    "messages": [{"role": "user", "content": prompt}],  # 역할 + 내용 쌍
    "temperature": 0.7,                                 # 응답의 무작위성 조절
    "max_tokens": 500,                                  # 최대 응답 길이
}
resp = requests.post(API_URL, headers=headers, json=body, timeout=30)
```

| 필드 | 타입 | 의미 |
| --- | --- | --- |
| `model` | `str` | 사용할 모델 (`"some-model"`, `"some-model"`) |
| `messages` | `list` | `role`,`content` 쌍 목록. 대화 이력을 순서대로 쌓음 |
| `temperature` | `float` | 낮을수록 일관된 답, 높을수록 무작위성이 큰 답 |
| `max_tokens` | `int` | 최대 응답 토큰 |

응답도 JSON입니다. 키와 인덱스에 순서대로 접근해 답변을 추출합니다.

```python
answer = result["choices"][0]["message"]["content"]
# dict 의 "choices" , list 의 0번 , dict 의 "message" , dict 의 "content"
```

| 필드 경로 | 의미 | 타입 |
| --- | --- | --- |
| `choices[0]["message"]["content"]` | 응답 텍스트 | `str` |
| `usage["total_tokens"]` | 사용 토큰 수 | `int` |
| `choices[0]["finish_reason"]` | 종료 이유 (`stop` 정상, `length` 길이초과) | `str` |

`choices` 는 `list` 이므로 `choices[0]` 으로 첫 응답을 꺼냅니다. 응답이 비면 `IndexError` 가 날 수 있습니다.

> **API 키 보안**: 실제 호출에는 인증용 **API 키** 가 필요합니다. API 키는 인증 정보이므로 코드나 공개 저장소에 올리지 않습니다. **환경 변수**(`os.environ`)로 불러오고, 노출된 키는 즉시 폐기한 뒤 다시 발급해야 합니다.

```python
import os

API_KEY = os.environ.get("OPENAI_API_KEY")   # 환경 변수에서 API 키 읽기
if not API_KEY:
    raise RuntimeError("OPENAI_API_KEY 환경 변수가 설정되지 않음")

headers = {"Authorization": f"Bearer {API_KEY}"}
```

터미널에서 `export OPENAI_API_KEY=...`(macOS,Linux) 또는 `setx`(Windows)로 등록합니다. `.env` 파일을 사용한다면 `.gitignore`에 `.env`를 추가합니다.
