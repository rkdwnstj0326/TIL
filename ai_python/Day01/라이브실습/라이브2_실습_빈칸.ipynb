## 리스트와 NumPy 배열

- 리스트에 숫자를 곱하면 각 원소를 계산하지 않고 리스트를 반복함
- 배열(ndarray) : 같은 자료형의 값을 담는 다차원 배열 구조
- 벡터화(Vectorization) : 반복문 없이 배열 전체에 연산을 적용


```python
가격_리스트 = [4000, 4500, 5000]
print(가격_리스트 * 2)            # [4000, 4500, 5000, 4000, 4500, 5000], 리스트를 두 번 반복

가격 = np.array([4000, 4500, 5000])
# 배열에 0.9 를 곱해 10% 할인가를 구해 보세요
할인가 = None                     # 가격 * 0.9
print(할인가)                     # [3600. 4050. 4500.]
```

## 배열 만들기

- `np.array(리스트)` : 리스트를 NumPy 배열로 변환
- `np.arange(n)` : 0부터 n-1까지의 정수로 배열을 생성
- `np.zeros(n)` : 0으로 채운 배열을 생성
- `np.linspace(a, b, n)` : a부터 b까지 같은 간격의 값 n개를 생성


```python
print(np.arange(5))             # [0 1 2 3 4]
print(np.zeros(3))              # [0. 0. 0.]

# 0 부터 1 까지 같은 간격으로 5 개의 값을 만들어 보세요
구간 = None                       # np.linspace(0, 1, 5)
print(구간)                       # [0.   0.25 0.5  0.75 1.  ]
```

## 벡터 연산과 통계

- 크기가 같은 배열끼리 연산하면 같은 위치의 원소끼리 계산됨
- `.sum()`, `.mean()`, `.max()` : 합계, 평균, 최댓값을 계산


```python
단가 = np.array([4000, 4500, 5000, 5500, 6000])
수량 = np.array([2, 1, 3, 5, 1])

# 같은 위치의 단가와 수량을 곱해 매출을 구해 보세요
매출 = None                       # 단가 * 수량
print(매출)                       # [ 8000  4500 15000 27500  6000]
print(매출.sum())                 # 61000, 합계
print(매출.mean())                # 12200.0, 평균
```

## 조건 필터와 정렬

- `배열[조건]` : 조건이 `True`인 위치의 원소만 추출
- `np.sort(배열)` : 오름차순으로 정렬한 새 배열을 반환
- `배열[::-1]` : 배열의 순서를 뒤집음


```python
매출 = np.array([8000, 4500, 15000, 25000, 6000])

# 1만 이상인 매출만 골라 보세요
고매출 = None                      # 매출[매출 >= 10000]
print(고매출)                      # [15000 25000]

print(np.sort(매출))               # [ 4500  6000  8000 15000 25000], 오름차순
print(np.sort(매출)[::-1])         # [25000 15000  8000  6000  4500], 내림차순
```

## reshape와 축 연산

- `arr.reshape(행, 열)` : 원소 수를 유지하면서 배열의 모양을 변경
- `axis=0` : 열별 집계
- `axis=1` : 행별 집계


```python
arr = np.arange(6)               # [0 1 2 3 4 5]

# arr 을 2 행 3 열로 바꿔 보세요
표 = None                          # arr.reshape(2, 3)
print(표)                          # [[0 1 2]\n [3 4 5]]
print(표.sum(axis=0))             # [3 5 7], 열별 합계
print(표.sum(axis=1))             # [3 12], 행별 합계
```

## DataFrame

- DataFrame : 행과 열로 구성된 2차원 표
- 컬럼마다 다른 자료형을 사용할 수 있음

`seed`를 고정했으므로 셀을 다시 실행해도 같은 데이터가 만들어집니다.


```python
np.random.seed(42)
n = 200
지역 = np.random.choice(["서울", "부산", "대전", "광주"], n)
메뉴 = np.random.choice(["아메리카노", "카페라떼", "바닐라라떼"], n)
단가 = np.random.choice([4000, 4500, 5000, 5500, 6000], n)
수량 = np.random.randint(1, 6, n)

df = pd.DataFrame({"지역": 지역, "메뉴": 메뉴, "단가": 단가, "수량": 수량})
df["매출"] = df["단가"] * df["수량"]   # 단가와 수량을 곱한 파생 컬럼

print(df.shape)                  # (200, 5)
df.head()
```

## DataFrame 메서드

- `df.shape` : 행 수와 열 수를 튜플로 반환
- `df.head(n)` : 위에서 n개 행을 반환
- `df.describe()` : 숫자 컬럼의 요약 통계를 반환
- `df.info()` : 컬럼, 자료형, 결측값을 확인


```python
행수, 열수 = df.shape
print("행:", 행수, "| 열:", 열수)   # 행: 200 | 열: 5

df.info()                        # 컬럼, 자료형, 결측 요약

# 숫자 컬럼의 요약 통계를 확인해 보세요
요약 = None                        # df.describe()
요약
```

## 조건 필터

- `df[조건]` : 조건이 `True`인 행만 반환
- 여러 조건은 `&`와 `|`로 연결하고 각 조건을 괄호로 감쌈
- Pandas 조건식에서는 `&`, `|`를 사용함 (`and`, `or` 사용 불가)


```python
# 지역이 "서울" 인 행만 골라 보세요
서울 = None                        # df[df["지역"] == "서울"]
print("서울 주문 수:", len(서울))

# 매출 1만 이상이면서 메뉴가 아메리카노인 행 (괄호 필수)
고매출아메 = df[(df["매출"] >= 10000) & (df["메뉴"] == "아메리카노")]
print("조건에 맞는 주문 수:", len(고매출아메))
서울.head()
```
