````python
import requests
import os
from dotenv import load_dotenv

# .env 파일 읽기
load_dotenv()

API_KEY = os.getenv("OPENAI_API_KEY")

if not API_KEY:
    raise RuntimeError("OPENAI_API_KEY가 .env 파일에 없습니다.")
else:
    print("OPENAI_API_KEY 읽기 성공")
    print(API_KEY[:10] + "...")
````

#### GPT-4 계열
 - Chat Completions API Reference : https://developers.openai.com/api/reference/resources/chat

````python
headers = {
    "Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}",
    "Content-Type": "application/json",
}
prompt = "대전의 유명한 관광지를 소개해줘."
API_URL = "https://api.openai.com/v1/chat/completions"

body = {
    "model": "gpt-4.1-mini",
    "messages": [
        {
            "role": "user",
            "content": prompt,
        }
    ],
    "temperature": 0.7,
    "max_tokens": 500,
}
response = requests.post(
    API_URL,
    headers=headers,
    json=body,
    timeout=30,
)
print(response.status_code)
print(response.json())
# print(response.text)

answer = response.json()["choices"][0]["message"]["content"]
print(answer)
````

#### GPT-5 계열
 - Responses API Reference : https://developers.openai.com/api/reference/resources/responses

````python
headers = {
    "Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}",
    "Content-Type": "application/json",
}
prompt = "대전의 유명한 맛집 소개해줘."
API_URL = "https://api.openai.com/v1/responses"

body = {
    "model": "gpt-5-mini",
    "input": prompt,
}
response = requests.post(
    API_URL,
    headers=headers,
    json=body,
    # timeout=30,
)
print(response.status_code)
# print(response.json())
print(response.text)

answer = response.json()["output"][1]["content"][0]["text"]
print(answer)
````
