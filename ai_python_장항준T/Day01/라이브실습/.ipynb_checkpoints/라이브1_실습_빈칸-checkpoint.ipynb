## 실습 방법

- 코드 셀의 `None`을 알맞은 값이나 코드로 바꾼 뒤 실행합니다.
- 오른쪽 주석에서 예상 결과를 확인할 수 있습니다.
- 셀은 위에서부터 순서대로 실행합니다.
- 정답 노트북은 강사용이며 실습을 마친 뒤 함께 확인합니다.


## 변수와 자료형

- 변수 : 데이터를 저장하는 공간
- `=` : 오른쪽 값을 왼쪽 변수에 담는 연산
- `type()` : 변수의 자료형을 확인
- 기본 자료형 : `int`(정수), `float`(실수), `str`(문자열), `bool`(참, 거짓)


```python
# 각 변수에 알맞은 값을 담고 실행해 보세요
age = None          # 정수 27
height = None       # 실수 175.5
name = None         # 문자열 "김싸피"
is_student = None   # 참 True

print(age, type(age))            # 27 <class 'int'>
print(height, type(height))      # 175.5 <class 'float'>
print(name, type(name))          # 김싸피 <class 'str'>
print(is_student, type(is_student))  # True <class 'bool'>
```

## 사칙연산

- `//` : 몫
- `%` : 나머지
- `**` : 제곱
- `/` : 나누기, 결과는 실수로 반환


```python
age = 27
print(age + 3)     # 30
print(age / 2)     # 13.5, 나누기 결과는 실수
print(age // 2)    # 13, 몫
print(age % 2)     # 1, 나머지

# age의 제곱을 구해 보세요
제곱 = None          # age ** 2
print(제곱)          # 729
```

## 형 변환과 f-string

- `int()`, `float()` : 문자열을 숫자로 변환
- `int()`로 실수를 정수로 변환하면 소수점 아래를 버림
- `str()` : 숫자를 문자열로 변환
- f-string : 문자열 안에 변수나 값을 넣는 표기


```python
# 문자열 "4000" 을 정수로 바꿔 보세요
price = None         # int("4000")
print(price, type(price))   # 4000 <class 'int'>

print(int(3.99))     # 3, 소수점 아래를 버림

count = 27
# f-string으로 "김싸피님은 27살" 을 만들어 보세요
문장 = None            # f"김싸피님은 {count}살"
print(문장)            # 김싸피님은 27살
```

## 리스트

- 리스트(list) : 여러 값을 순서대로 담는 자료구조
- 인덱스(index) : 각 값에 접근하기 위한 번호로 0부터 시작함
- 슬라이싱 `[시작:끝]` : 여러 값을 잘라서 가져오며 끝 인덱스는 포함하지 않음


```python
scores = [90, 85, 70, 60, 50]
print(scores[0])    # 90, 첫 번째
print(scores[-1])   # 50, 마지막

# 인덱스 1부터 3까지 잘라 보세요 (끝 인덱스 4는 제외)
가운데 = None          # scores[1:4]
print(가운데)          # [85, 70, 60]
```

## 리스트 메서드

- `append(x)` : 리스트 끝에 값을 추가
- `pop()` : 값을 제거한 후 제거한 값을 반환
- `sort()` : 기존 리스트를 직접 정렬


```python
scores = [90, 85, 70]
scores.append(60)     # 마지막에 60 추가
print(scores)         # [90, 85, 70, 60]

# 마지막 값을 제거하고 반환값을 받아 보세요
꺼낸값 = None           # scores.pop()
print(꺼낸값, scores)   # 60 [90, 85, 70]
```

## 딕셔너리

- 딕셔너리(dict) : 키(key)와 값(value)을 쌍으로 담는 자료구조
- `dict["키"]` : 키로 값에 접근하며 키가 없으면 `KeyError`가 발생함
- `get("키", 기본값)` : 키가 없으면 지정한 기본값을 반환


```python
person = {"name": "김싸피", "age": 27}
print(person["name"])            # 김싸피

# 없는 키 "hobby" 를 기본값 "없음" 으로 꺼내 보세요
취미 = None                        # person.get("hobby", "없음")
print(취미)                        # 없음

person["job"] = "학생"           # 키가 없으면 추가
print(person)                     # {'name': '김싸피', 'age': 27, 'job': '학생'}
```

## 리스트와 딕셔너리의 중첩

- 리스트 안에 딕셔너리, 딕셔너리 안에 리스트를 넣을 수 있음
- 리스트는 인덱스, 딕셔너리는 키로 값에 접근함
- 중첩된 값은 바깥쪽 구조부터 차례대로 접근함


```python
students = [
    {"name": "김싸피", "scores": [90, 85, 70]},
    {"name": "민지", "scores": [80, 95, 100]},
]

# 인덱스 0 에 있는 학생의 이름을 꺼내 보세요
이름 = None            # students[0]["name"]
print(이름)            # 김싸피

# 인덱스 1 에 있는 학생의 첫 점수를 꺼내 보세요
첫점수 = None          # students[1]["scores"][0]
print(첫점수)          # 80
```

## 조건문

- 조건의 참, 거짓에 따라 다른 코드를 실행함
- 위에서부터 조건을 확인해 처음으로 참인 분기만 실행함
- `==` : 두 값이 같은지 비교


```python
score = 75

# score가 60 이상이면 "합격" 을 출력하도록 빈칸을 채워 보세요
if score >= 90:
    print("A")
elif score >= None:      # 60
    print("합격")
else:
    print("불합격")
# 출력: 합격
```

## 반복문

- `for` : 묶음에서 값을 하나씩 꺼내며 반복
- `range(n)` : 0부터 n-1까지의 숫자를 생성


```python
# 점수를 모두 더해 합계를 구합니다
scores = [90, 85, 70, 60]
total = 0
for s in scores:
    total = total + s      # 각 점수를 total에 더함
print(total)               # 305

# range로 0, 1, 2 를 순서대로 출력해 보세요
for i in range(None):       # range(3)
    print(i)                # 0, 1, 2
```

## API 요청과 requests

- `requests.get(URL)` : 지정한 URL에 GET 요청을 보내고 응답 객체를 반환
- `status_code` : 요청 결과를 나타내는 상태 코드
- `.json()` : JSON 응답을 파이썬 객체로 변환

네트워크 연결에 실패하면 같은 구조의 샘플 데이터를 사용합니다. 아래 셀은 수정하지 않고 실행합니다.


```python
import requests

URL = "https://jsonplaceholder.typicode.com/users"
SAMPLE = [
    {"name": "Leanne Graham", "email": "Sincere@april.biz",
     "address": {"city": "Gwenborough"}},
    {"name": "Ervin Howell", "email": "Shanna@melissa.tv",
     "address": {"city": "Wisokyburgh"}},
]

try:
    response = requests.get(URL, timeout=5)
    print("상태 코드:", response.status_code)   # 200, 요청 성공
    data = response.json()                       # 응답을 파이썬 객체로 변환
except Exception as e:
    print("네트워크를 사용할 수 없어 샘플 데이터로 진행:", e)
    data = SAMPLE

print("사용자 수:", len(data))
```

## JSON 데이터 접근

- 리스트는 인덱스, 딕셔너리는 키로 값에 접근함
- 중첩된 데이터는 바깥쪽 구조부터 차례대로 접근함


```python
first = data[0]                 # 첫 번째 사용자
print(first["name"])            # 이름
print(first["email"])           # 이메일

# 첫 번째 사용자의 도시를 꺼내 보세요 (address의 city)
도시 = None                       # first["address"]["city"]
print(도시)
```
